# Logical Operators in Python

* Logical operators in Python are used to combine multiple conditions and return a Boolean result: True or False.

* They are very important in Python, Data Analytics, Machine Learning, and interview preparation because they help us filter data and make decisions.

# 1. Different logical operators


# Operator                  Name             Meaning

* and                   Logical AND      Both conditions must be True

* or                    Logical OR       At least one condition must be True

* not                   Logical NOT       Reverses the result

```python
# Example 1: Logical AND

# The and operator returns True only when both conditions are True.

age = 25
salary = 50000

print(age > 18 and salary > 30000)
```

```python
# Example 2: Logical OR

# The or operator returns True when at least one condition is True.

age = 16
salary = 50000

print(age > 18 or salary > 30000)
```

```python
# Example 3: Logical NOT

# The not operator reverses a Boolean value.

is_student = True

print(is_student)
print(not is_student)
```

```python
# Example 4: Logical operators with if-else

# This is a practical example of checking whether a person is eligible for a loan.

age = 25
salary = 50000

if age >= 21 and salary >= 30000:
    print("Eligible for loan")
else:
    print("Not eligible for loan")
```

```python
## Example 5: Logical operators with Data Analytics

### Suppose we have employee data. We want to check whether an employee is eligible for a bonus.

## Conditions:

# Salary is greater than 40,000.

# Experience is greater than 3 years.

salary = 50000
experience = 5

if salary > 40000 and experience > 3:
    print("Eligible for bonus")
else:
    print("Not eligible for bonus")
```

```python
# Example using OR

# An employee can get a bonus if they satisfy either condition:

salary = 30000
experience = 5

if salary > 40000 or experience > 3:
    print("Eligible for bonus")
else:
    print("Not eligible for bonus")
```

```python
# Using parentheses

# Parentheses help control the order in which conditions are evaluated.

age = 25
salary = 20000
is_employee = True

result = (age > 18 and salary > 30000) or is_employee

print(result)
```

```python

# 8. Logical operators with strings

# Logical operators can also be used with strings.

username = "admin"
password = "1234"

if username == "admin" and password == "1234":
    print("Login successful")
else:
    print("Invalid username or password")
```

```python
# 9. Logical operators with lists

# You can check whether a value exists in a list using in.

skills = ["Python", "SQL", "Machine Learning"]

print("Python" in skills)
print("Java" in skills)
```

```python
skills = ["Python", "SQL", "Machine Learning"]

if "Python" in skills and "SQL" in skills:
    print("Good for Data Analyst role")
else:
    print("Need to improve skills")
```

# String Methods in Python

* A string is a sequence of characters written inside quotes.

## name = "Suraj"

* Python provides many built-in string methods to modify, search, and analyze strings.

* Important: Strings are immutable, meaning string methods return a new string. They do not change the original string directly.

```python
# 1. Common String Methods

# Method                          Purpose                          Example

# upper()                   Converts to uppercase         "python".upper() → "PYTHON"

# lower()                   Converts to lowercase         "PYTHON".lower() → "python"

# title()           First letter of each word uppercase    "hello world".title()

# capitalize()           First character uppercase          "python".capitalize()

# strip()              Removes spaces from both ends         " hello ".strip()

# replace()                   Replaces text                 "Hello".replace("H", "J")

# split()              Converts string into a list           "a,b,c".split(",")

# join()               Joins list items into a string        "-".join(["a","b"])

# find()                Finds position of text               "Python".find("t")

# count()               Counts occurrences                    "banana".count("a")

# startswith()         Checks starting text                  "Python".startswith("Py")

# endswith()               Checks ending text                  "Python".endswith("on")

# isdigit()       Checks whether all characters are digits      "123".isdigit()

# isalpha() Checks whether all characters are alphabets          "Python".isalpha()

# isalnum()          Checks alphabets and numbers              "Python123".isalnum()
```

```python
# Example 1: Case Conversion Methods

# Methods used:

# upper()

# lower()

# title()

# capitalize()

text = "python programming language"

print("Original:", text)
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Title case:", text.title())
print("Capitalized:", text.capitalize())
```

```python
# Example 2: Removing and Replacing Text

# Methods used:

# strip()

# lstrip()

# rstrip()

# replace()
```

```python
text = "   Hello Python   "

print("Original:", repr(text))
print("After strip:", repr(text.strip()))
print("Left strip:", repr(text.lstrip()))
print("Right strip:", repr(text.rstrip()))

message = "I love Java"
print("Before replace:", message)
print("After replace:", message.replace("Java", "Python"))
```

```python
# Example 3: Searching and Counting Characters

# Methods used:

# find()

# index()

# count()

# startswith()

# endswith()

text = "python programming"

print("Position of 'p':", text.find("p"))
print("Position of 'programming':", text.find("programming"))
print("Count of 'p':", text.count("p"))
print("Starts with python:", text.startswith("python"))
print("Ends with programming:", text.endswith("programming"))
```

```python
# Example 4: Split and Join Methods

# Methods used:

# split()

# join()

sentence = "Python is easy to learn"

words = sentence.split()

print("Original sentence:", sentence)
print("After split:", words)
print("Number of words:", len(words))

new_sentence = "-".join(words)

print("After join:", new_sentence)
```

```python
# Example 5: Checking String Content

# Methods used:

# isalpha()

# isdigit()

# isalnum()

# isspace()

# islower()

# isupper()
```

```python
values = ["Python", "12345", "Python123", " ", "python", "PYTHON"]

for value in values:
    print("Value:", repr(value))
    print("Alphabet only:", value.isalpha())
    print("Digit only:", value.isdigit())
    print("Alphanumeric:", value.isalnum())
    print("Space only:", value.isspace())
    print("Lowercase:", value.islower())
    print("Uppercase:", value.isupper())
    print("-" * 30)
```

```python
# Practical Example: Cleaning Data Using String Methods

# This is useful in Data Analytics and Data Science.

names = ["  suraj ", " PRIYA", "rahul  ", "  amit"]

clean_names = []

for name in names:
    cleaned_name = name.strip().title()
    clean_names.append(cleaned_name)

print(clean_names)
```

```python
# Practical Example: Email Validation
email = "suraj@gmail.com"

if email.endswith("@gmail.com"):
    print("This is a Gmail address")
else:
    print("This is not a Gmail address")
```

```python
# Practical Example: Count Words in a Sentence

sentence = "Python is easy and Python is powerful"

words = sentence.lower().split()

print("Words:", words)
print("Total words:", len(words))
print("Python count:", words.count("python"))
```

# String Indexing in Python

## String indexing means accessing individual characters from a string using their position (index).

### Important rule

### Python indexing starts from 0, not 1.

#### For example:

```python
text = "   Hello, World!   "
```

```python


# Character:   P   y   t   h   o   n
# Positive:    0   1   2   3   4   5
# Negative:   -6  -5  -4  -3  -2  -1
```

```python
# 1. Positive Indexing

# Positive indexing starts from left to right, beginning with 0.

text = "Python"

print(text[0])
print(text[1])
print(text[2])
print(text[5])
```

```python
# 2. Negative Indexing

# Negative indexing starts from the right side.

text = "Python"

print(text[-1])
print(text[-2])
print(text[-3])
print(text[-6])
```

```python
# 3. String Slicing

# Slicing is used to get a part of a string.

# Syntax:

# string[start:stop]

# The start index is included, but the stop index is not included.
```

```python
text = "Python"

print(text[0:2])
print(text[0:4])
print(text[2:5])
```

```python
# 4. Slicing with Start and Stop

# You can omit the start or stop position.

text = "Python Programming"

print(text[:6])
print(text[7:])
print(text[:])
```

```python
# 5. Slicing with Step

# The complete slicing syntax is:

# string[start:stop:step]
```

```python
text = "Python"

print(text[::1])
print(text[::2])
print(text[::3])
```

```python
# 6. Reverse a String

# One of the most important uses of string indexing is reversing a string.

# Use:

# [::-1]
```

```python
text = "Python"

reverse_text = text[::-1]

print(reverse_text)
```

```python
#Practical Example: Extract First and Last Character
name = "Suraj"

first = name[0]
last = name[-1]

print("First character:", first)
print("Last character:", last)
```

```python
#Practical Example: Extract a Part of an Email
email = "suraj@gmail.com"

username = email[:email.find("@")]
domain = email[email.find("@") + 1:]

print("Username:", username)
print("Domain:", domain)
```
