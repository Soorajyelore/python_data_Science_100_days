# Format Specifiers in Python

A format specifier is used to control how a value is displayed when we print or format a string.

Format specifiers are commonly used with f-strings and the .format() method.

# For example:

```python
name = "Suraj"
age = 35

print(f"My name is {name} and I am {age} years old.")
```

When you want to control the number of decimal places, alignment, percentage, commas, etc., format specifiers become very useful.

# 1. Basic Syntax

* With an f-string:

* f"{value:format_specifier}"

# Example:

```python
price = 125.6789

print(f"{price:.2f}")
```

# Here:

* price → value
* : → starts the format specification
* .2f → display 2 decimal places

# Important Format Specifiers

| Specifier | Meaning             | Example       |
| --------- | ------------------- | ------------- |
| `d`       | Integer             | `{num:d}`     |
| `f`       | Floating-point      | `{num:f}`     |
| `.2f`     | 2 decimal places    | `{num:.2f}`   |
| `%`       | Percentage          | `{value:.2%}` |
| `,`       | Thousands separator | `{salary:,}`  |
| `e`       | Scientific notation | `{num:e}`     |
| `s`       | String              | `{name:s}`    |
| `>`       | Right alignment     | `{name:>10}`  |
| `<`       | Left alignment      | `{name:<10}`  |
| `^`       | Center alignment    | `{name:^10}`  |
| `0`       | Zero padding        | `{num:05d}`   |


# Example 1: Decimal Places

* This is one of the most common uses.

```python
price = 125.6789

print(f"Original price: {price}")
print(f"Price: {price:.2f}")
print(f"Price: {price:.3f}")
```

# Example 2: Percentage Formatting

* Suppose a model has an accuracy of 0.9567.

```python
accuracy = 0.9567

print(f"Model Accuracy: {accuracy:.2%}")
```

# Python converts:

0.9567 → 95.67%

# Data Science Example


```python
accuracy = 0.9235
precision = 0.8912
recall = 0.8756

print(f"Accuracy : {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall   : {recall:.2%}")
```

# Example 3: Thousands Separator

* Suppose you have a large salary or revenue value.

```python
salary = 1250000

print(f"Salary: {salary:,}")
```

# Another example

```python
revenue = 125678900

print(f"Revenue: ₹{revenue:,}")
```

# Example 4: Alignment

* Format specifiers can control how text is positioned.

* Left Alignment

```python
name = "Python"

print(f"{name:<15}")# The text is aligned to the left within 15 spaces.
```

# Center Alignment

```python

name = "Python"

print(f"{name:^15}")

```

# Example 5: Zero Padding

* Zero padding is useful when you want a fixed number of digits.

```python
number = 25

print(f"{number:05d}")
```

# means:

* 0 → fill with zeros
* 5 → total width of 5 characters
* d → integer

```python
employee_id = 42

print(f"EMP{employee_id:05d}")
```

# Example 6: Scientific Notation

* Scientific notation is useful for very large or very small numbers.

```python
number = 123456789

print(f"{number:e}")
```

# Example 7: Combining Multiple Format Specifiers

* You can combine formatting options.

```python
price = 1234567.89123

print(f"Price: ₹{price:,.2f}")



# Here:

# ,   → thousands separator
# .2  → 2 decimal places
# f   → floating-point number

# This is a very useful format in real-world applications.
```

# Example 8: Creating a Simple Data Report

* You can combine several format specifiers to create a clean report.

```python
name = "Suraj"
salary = 1250000
accuracy = 0.9456
experience = 5.5

print(f"Name       : {name}")
print(f"Salary     : ₹{salary:,}")
print(f"Experience : {experience:.1f} years")
print(f"Accuracy   : {accuracy:.2%}")
```

# Common Format Specifiers to Remember


| Format | Example     | Result         |
| ------ | ----------- | -------------- |
| `:.2f` | `125.6789`  | `125.68`       |
| `:.3f` | `125.6789`  | `125.679`      |
| `:.2%` | `0.9567`    | `95.67%`       |
| `:,`   | `1250000`   | `1,250,000`    |
| `:05d` | `25`        | `00025`        |
| `:<10` | `"Python"`  | Left aligned   |
| `:>10` | `"Python"`  | Right aligned  |
| `:^10` | `"Python"`  | Center aligned |
| `:.2e` | `123456789` | `1.23e+08`     |


# f-string vs .format()
* Using f-string

```python
price = 125.6789

print(f"Price: ₹{price:.2f}")
# Using .format()
price = 125.6789

print("Price: ₹{:.2f}".format(price))

# Both produce:

# Price: ₹125.68

# For modern Python code, f-strings are generally the easiest and most readable approach.
```

# Interview Answer

Format specifiers in Python are used to control how values are displayed when formatting strings. They are commonly used with f-strings. For example, .2f displays two decimal places, ,.2f adds thousands separators and two decimal places, % formats a value as a percentage, and <, >, and ^ control text alignment.
