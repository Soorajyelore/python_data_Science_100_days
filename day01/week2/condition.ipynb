# Conditional Expressions in Python

* A conditional expression is a short way to make a decision and return a value based on a condition.

# The most common syntax is:

* value_if_true if condition else value_if_false

# For example:



```python
age = 20

result = "Adult" if age >= 18 else "Minor"

print(result)
```

# Conditional Expression Types — Quick Summary


| Type       | Example                                          | Use                  |
| ---------- | ------------------------------------------------ | -------------------- |
| Simple     | `x if condition else y`                          | Two choices          |
| Numeric    | `a if a > b else b`                              | Calculations         |
| Logical    | `"Yes" if age > 18 and salary > 50000 else "No"` | Multiple conditions  |
| Nested     | `"A" if x > 90 else "B" if x > 70 else "C"`      | Multiple outcomes    |
| User input | `"Pass" if marks >= 40 else "Fail"`              | Interactive programs |


# Interview Answer

A conditional expression in Python is a concise way to write an if-else decision in a single line. Its syntax is value_if_true if condition else value_if_false. It is also known as a ternary conditional expression and is useful when the logic is simple and readable.
