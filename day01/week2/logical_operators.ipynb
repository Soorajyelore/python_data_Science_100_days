# Logical Operators in Python

* Logical operators are used to combine multiple conditions and return either True or False.

* Python has 3 main logical operators:

| Operator | Meaning                               | Example                   |
| -------- | ------------------------------------- | ------------------------- |
| `and`    | Both conditions must be `True`        | `age >= 18 and age <= 60` |
| `or`     | At least one condition must be `True` | `age < 18 or age > 60`    |
| `not`    | Reverses the result                   | `not(age >= 18)`          |


# 1. and Operator

* The and operator returns True only when both conditions are True.

# Example 1: Age and Experience

```python
age = 25
experience = 3

if age >= 18 and experience >= 2:
    print("Eligible")
else:
    print("Not Eligible")
```

# 2. or Operator

* The or operator returns True when at least one condition is True.

# Example 2: Discount Eligibility

```python
age = 65
premium_member = False

if age >= 60 or premium_member:
    print("Discount available")
else:
    print("No discount")
```

# 3. not Operator

* The not operator reverses the result.

# Example 3

```python
age = 15

if not age >= 18:
    print("You are under 18")
else:
    print("You are 18 or older")
```

# 4. Using and, or Together

* You can combine multiple logical operators.

# Example 4: Job Eligibility

```python
age = 28
experience = 4
degree = True

if age >= 18 and experience >= 2 and degree:
    print("Eligible for the job")
else:
    print("Not eligible")
```

# 5. Practical Example with User Input

* This is a good example for practice.

```python
marks = int(input("Enter your marks: "))
attendance = int(input("Enter your attendance: "))

if marks >= 40 and attendance >= 75:
    print("Student is eligible for the exam")
else:
    print("Student is not eligible")
```

# Interview Answer

Logical operators in Python are used to combine or reverse conditions. Python provides three logical operators: and, or, and not. and requires all conditions to be true, or requires at least one condition to be true, and not reverses the Boolean result.
