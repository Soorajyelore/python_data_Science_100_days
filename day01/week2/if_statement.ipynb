 # if statements

# If Statements in Python

* An if statement is used to make decisions in a Python program.

* It checks whether a condition is True or False.

# Basic Syntax

* if condition:
    # code to execute

# Example:



```python
age = 20

if age >= 18:
    print("You are eligible to vote")
```

# Different Types of if Statements

* There are mainly 5 types you should learn:

* 1.Simple if
* 2.if-else
* 3.if-elif-else
* 4.Nested if
* 5.Multiple conditions using and / or

# 1. Simple if Statement

* The code runs only when the condition is True.

# Example 1: Check Age

```python
age = 25

if age >= 18:
    print("You are an adult")
```

# if age is 15, nothing will be printed because the condition is False.

*Flow
* Condition
*   ↓
* True → Execute code
* False → Skip code

# 2. if-else Statement

* Use if-else when there are two possible outcomes.

# Example 2: Pass or Fail

```python
marks = 65

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

# Structure
*if condition:
*    # True
* else:
*    # False

# 3. if-elif-else Statement

* Use elif when you have multiple conditions.

# Example 3: Grade Calculation

```python
marks = 78

if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```

# 4. Nested if Statement

* An if statement inside another if statement is called a nested if.

# Example 4: Login System

```python
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Login successful")
    else:
        print("Wrong password")
else:
    print("Wrong username")
```

# 5. if with Multiple Conditions

#### You can combine multiple conditions using:

* and
* or
* not
# Example 5: Job Eligibility

```python
age = 25
experience = 3

if age >= 18 and experience >= 2:
    print("Eligible for the job")
else:
    print("Not eligible")
```

# Example 6: Using or

* Suppose a customer gets a discount if they are either a senior citizen or a premium member.

```python
age = 30
premium_member = True

if age >= 60 or premium_member:
    print("Discount available")
else:
    print("No discount")
```

# Example 7: User Input + if-elif-else

* This is a very important practical example.

```python
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Excellent")
elif marks >= 75:
    print("Very Good")
elif marks >= 60:
    print("Good")
elif marks >= 40:
    print("Pass")
else:
    print("Fail")
```

# Example 8: Data Science Example

* Suppose we want to classify a customer's age group.

```python
age = 35

if age < 18:
    category = "Minor"
elif age < 30:
    category = "Young Adult"
elif age < 60:
    category = "Adult"
else:
    category = "Senior Citizen"

print("Category:", category)
```

# Quick Comparison


| Type                | When to use                        | Example         |
| ------------------- | ---------------------------------- | --------------- |
| `if`                | One condition                      | Age check       |
| `if-else`           | Two possibilities                  | Pass/Fail       |
| `if-elif-else`      | Multiple possibilities             | Grades          |
| Nested `if`         | Condition inside another condition | Login           |
| Multiple conditions | Combine conditions                 | Job eligibility |


Interview Answer

An if statement in Python is a decision-making statement used to execute a block of code when a condition is true. Python supports simple if, if-else, if-elif-else, nested if, and conditions combined using logical operators such as and, or, and not.
