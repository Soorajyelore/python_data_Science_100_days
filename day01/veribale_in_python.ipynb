# Variables in Python

* A variable in Python is a name used to store a value. We use variables to store data such as numbers, text, and lists.

# For example:

```python
name = "Suraj"
age = 35
salary = 50000
```

# Here:

### name is a variable.

* "Suraj" is the value.

* age is a variable.

* 35 is the value.

# Python automatically identifies the data type of a variable.

# 1. Types of Variables in Python

## Python has several built-in data types. The most common types are:

### Data type             Meaning                           Example

*     int           Integer / whole number                 age = 35

*    float             Decimal number                        salary = 50000.50

*     str              String / text                      name = "Suraj"

*     bool             True or False                 is_active = True

*     list        Collection of values                   marks = [80, 90, 70]

*     tuple    Ordered, immutable collection       coordinates = (10, 20)

*     dict             Key-value pairs                 student = {"name": "Suraj"}

*     set              Unique values                   numbers = {1, 2, 3}

# Example 1: Integer (int)

* An integer is a whole number without a decimal point.

```python
age = 35
experience = 5
number_of_students = 100

print(age)
print(experience)
print(number_of_students)

print(type(age))
```

# Example 2: Float (float)

# A float is a number containing a decimal point.

```python
salary = 50000.50
height = 5.8
percentage = 85.5

print(salary)
print(height)
print(percentage)

print(type(salary))
```

# Example 3: String (str)

* A string is a collection of characters or text. We write strings inside single or double quotes.

```python
name = "Suraj"
city = "Pune"
job_role = "Data Scientist"

print(name)
print(city)
print(job_role)

print(type(name))
```

# Example 4: Boolean (bool)

* A Boolean variable stores only two values:

* True

* False

```python
is_student = True
is_employee = True
is_married = False

print(is_student)
print(is_employee)
print(is_married)

print(type(is_student))
```

# Example 5: List (list)

* A list stores multiple values in one variable. Lists are ordered and can be modified.

```python
marks = [80, 90, 70, 85, 95]

print(marks)
print(marks[0])
print(type(marks))
```

# Example 6: Dictionary (dict)

* A dictionary stores data in key-value pairs.

```python
student = {
    "name": "Suraj",
    "age": 35,
    "city": "Pune"
}

print(student)
print(student["name"])
print(student["age"])

print(type(student))


# Explanation:

# "name" is a key.

# "Suraj" is its value.

```

# How to check the type of a variable

* Use the built-in type() function.

```python
a = 10
b = 10.5
c = "Python"
d = True
e = [1, 2, 3]
f = {"name": "Suraj"}

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
```

# Important: Python is dynamically typed

* In Python, you do not need to declare the data type of a variable. Python identifies it automatically.

```python
x = 10

print(x)
print(type(x))

x = "Python"

print(x)
print(type(x))

x = 10.5

print(x)
print(type(x))
```

# Interview answer: What is a variable in Python?

* A variable in Python is a name used to store a value in memory. Python is dynamically typed, so we don't need to declare the data type explicitly. Python automatically identifies the type based on the assigned value. For example, age = 35 creates an integer variable, and name = "Suraj" creates a string variable.
