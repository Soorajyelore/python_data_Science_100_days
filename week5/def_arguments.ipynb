# default arguments

# 1. What are default arguments in Python?

- A default argument is a parameter that already has a value assigned in a function definition. If you call the function without providing that argument, Python uses the default value.


- Default arguments help you write reusable functions without passing the same values repeatedly.

# Basic syntax

```python
def greet(name="Guest"):
    print("Hello", name)

greet("Suraj")
greet()
```

#### How it works:
1. def greet(name="Guest") defines a function with a default value of "Guest".
2. greet("Suraj") passes a value explicitly, so Python uses "Suraj".
3. greet() does not pass a value, so Python uses "Guest".

### 2. Different types of function arguments
Python has several argument styles. Default arguments are one of them.

|Argument type	|Meaning	|Example
:---------------|-----------|------------
Positional arguments|	Values are matched by position	|add(10, 20)
Keyword arguments|	Values are matched by parameter name	|add(a=10, b=20)
Default arguments|	Parameters have predefined values	|greet(name="Guest")
Variable-length positional arguments (*args)|	Accept multiple positional values	|total(10, 20, 30)
Variable-length keyword arguments (**kwargs)|	Accept multiple named values	|show(name="Suraj", age=30)

# Example 1: Default argument with a name
Problem statement: Create a greeting function that says hello to a user. If the user does not provide a name, greet them as "Guest".

```python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet("Suraj")
greet("Amit")
greet()
```

* Step-by-step explanation:
1. greet() defines a function with one parameter, name.
2. "Guest" is the default value.
3. greet("Suraj") replaces the default with "Suraj".
4. greet("Amit") uses "Amit".
5. greet() uses "Guest" because no argument was supplied.

**Use case: Greeting users in applications, setting default names, and creating reusable functions.**

# Example 2: Multiple default arguments

**Problem statement:** Calculate the final price of a product. Use a default tax rate of 18% and a default discount of 0% if the caller does not specify them.

```python
def calculate_price(price, tax_rate=0.18, discount=0):
    tax = price * tax_rate
    discount_amount = price * discount
    final_price = price + tax - discount_amount

    return final_price

print(calculate_price(1000))
print(calculate_price(1000, 0.05))
print(calculate_price(1000, 0.18, 0.10))
```

### Step-by-step explanation:
1. price is required because it has no default value.
2. tax_rate=0.18 means the default tax rate is 18%.
3. discount=0 means no discount is applied by default.
4. tax = price * tax_rate calculates the tax.
5. discount_amount = price * discount calculates the discount.
6. return final_price sends the calculated result back to the caller.


### For the first call:
- Price = ₹1,000
- Tax = ₹180
- Discount = ₹0
- Final price = ₹1,180
* In the second call, 0.05 is passed as the tax rate, so the result is ₹1,050.
Use case: Pricing calculations, tax calculations, discounts, and financial applications.

### Example 3: Default arguments with keyword arguments


Problem statement: Create an employee information function with default department and city values. Allow the caller to override them by name.

```python
def employee_details(name, department="IT", city="Pune"):
    print("Name:", name)
    print("Department:", department)
    print("City:", city)

employee_details("Suraj")

print("---")

employee_details(
    "Amit",
    city="Mumbai",
    department="Data Science"
)
```

### Step-by-step explanation:
1. name is required.
2. department defaults to "IT".
3. city defaults to "Pune".
4. employee_details("Suraj") uses both defaults.
5. In the second call, keyword arguments explicitly override both defaults.
6. Keyword arguments can be provided in a different order because they are matched by name.

## Use case: Employee management systems, configuration functions, and report generation.

####  Example 4: Default argument for data analysis
* Problem statement: Calculate the average of a list of numbers. If no list is provided, use a predefined list of marks.

```python
def calculate_average(numbers=None):
    if numbers is None:
        numbers = [70, 80, 90]

    average = sum(numbers) / len(numbers)
    return average

print(calculate_average())
print(calculate_average([60, 75, 85, 100]))
```

* Step-by-step explanation:
1. The parameter numbers defaults to None.
2. If the caller supplies no list, the condition numbers is None is true.
3. Python assigns [70, 80, 90] as the data to analyze.
4. sum(numbers) calculates the total.
5. len(numbers) counts the values.
6. Dividing the total by the count gives the average.

* Why not write def calculate_average(numbers=[70, 80, 90])? Although that can work when the list is never modified, using None is a safer general pattern for mutable defaults such as lists and dictionaries.
* Use case: Data analytics, calculating average marks, sales analysis, and reusable statistical functions.

# Example 5: Default argument with *args
* Problem statement: Create a function that calculates the total of any number of values. Use a default discount of 0% and allow multiple prices to be passed.

```python
def calculate_total(*prices, discount=0):
    total = sum(prices)
    discount_amount = total * discount
    final_total = total - discount_amount

    return final_total

print(calculate_total(100, 200, 300))
print(calculate_total(100, 200, 300, discount=0.10))
```

Step-by-step explanation:
1. *prices collects any number of positional arguments into a tuple.
2. sum(prices) calculates the total.
3. discount=0 provides a default discount of 0%.
4. discount_amount calculates the discount on the total.
5. discount=0.10 applies a 10% discount when explicitly provided.


#### Notice that discount is written after *prices, so it must be passed by keyword in this function.
##### Use case: Shopping carts, billing systems, flexible mathematical functions, and sales reports.

### Example 6: Default arguments for a machine learning report

**Problem statement:** Create a reusable function to print a machine learning model's evaluation results. If no model name or accuracy is provided, use sensible defaults.

```python
def model_report(model_name="Random Forest", accuracy=0.90):
    print("Model:", model_name)
    print(f"Accuracy: {accuracy:.2%}")

model_report()

model_report("Logistic Regression", 0.85)

model_report(accuracy=0.92)
```

Step-by-step explanation:
1. model_name defaults to "Random Forest".
2. accuracy defaults to 0.90.
3. model_report() uses both default values.
4. model_report("Logistic Regression", 0.85) overrides both defaults.
5. model_report(accuracy=0.92) overrides only accuracy and keeps the default model name.
6. :.2% formats a decimal as a percentage with two digits after the decimal point.
Use case: Machine learning experiment reports, reusable evaluation functions, and model comparison summaries.
Note: Accuracy alone is not sufficient for evaluating every ML model, especially on imbalanced datasets.
