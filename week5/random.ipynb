# Random Numbers in Python

- In Python, random numbers are numbers generated by a program using a random-number generator. They are useful for simulations, games, sampling data, testing machine learning models, and creating random test data.

**Python** provides the built-in **random module** for generating random **integers**, **decimal numbers**, **selecting random items**, **shuffling lists**, and more.

# 1. How to use random numbers in Python
-First, import the random module.

```python
import random
```

|Function	|Purpose	|Example
:-----------|-----------|--------
|random.randint(a, b)	|Random integer, including both endpoints	|random.randint(1, 10)
|random.randrange(start, stop)	|Random integer; excludes stop	|random.randrange(1, 10)
|random.random()	|Random decimal from 0.0 up to, but not including, 1.0	|random.random()
|random.uniform(a, b)	|Random floating-point number in a range	|random.uniform(1, 10)
|random.choice(sequence)	|Select one item randomly	|random.choice(["A", "B"])
|random.sample(sequence, k)	|Select k unique positions/items without replacement	|random.sample([1, 2, 3, 4], 2)
|random.shuffle(list)	|Shuffle a list in place	|random.shuffle(numbers)
|random.seed(value)	|Initialize the generator for reproducible results	|random.seed(42)

#Example 1: Generate random integers
* Problem statement


-Generate a random integer between 1 and 100, like rolling a number for a simple game.

```python
import random

number = random.randint(1, 100)

print("Random number:", number)
```

### Step-by-step explanation
1. import random imports Python's random-number module.
2. random.randint(1, 100) generates an integer from 1 through 100, including both 1 and 100.
3. The generated number is stored in the number variable.
4. print() displays the result.

### Use cases: Dice games, lottery-style simulations, random IDs for practice,

# Example 2: Generate a random integer using randrange()

* Problem statement

- You want to randomly select an even number from 2 to 20.

```python
import random

number = random.randrange(2, 21, 2)

print("Random even number:", number)
```

## Step-by-step explanation
1. 2 is the starting value.
2. 21 is the stopping value, but it is excluded.
3. 2 is the step size, so the possible values are 2, 4, 6, ..., 20.
4. randrange() randomly selects one of those values.
* Use cases: Generating even or odd numbers, selecting numbers at fixed intervals, and testing algorithms with specific integer patterns.

Compare the two functions:

-random.randint(1, 10)   # Includes 10


-random.randrange(1, 10) # Excludes 10

# Example 3: Generate a random decimal using random()

- Problem statement

* You are simulating a probability value for an experiment. Generate a random decimal between 0 and 1.

```python
import random

value = random.random()

print("Random value:", value)
```

#### Step-by-step explanation
1. random.random() generates a floating-point number.
2. The number is greater than or equal to 0.0.
3. It is always less than 1.0.
4. Store and print the value.
##### For example, you can use the random value to simulate a simple 70% chance event:

```python
import random

value = random.random()

if value < 0.70:
    print("Event occurred")
else:
    print("Event did not occur")
```

The event has a 70% probability under this simulation.


Use cases: Probability experiments, Monte Carlo simulations, and randomized algorithm testing.

## Example 4: Generate a random decimal using uniform()
* Problem statement
* A temperature simulator needs a random decimal temperature between 20.0°C and 35.0°C.

```python
import random

temperature = random.uniform(20.0, 35.0)

print("Simulated temperature:", round(temperature, 2), "°C")
```

## Step-by-step explanation
1. uniform(20.0, 35.0) generates a floating-point number within the specified range.
2. The result is stored in temperature.
3. round(temperature, 2) rounds it to two decimal places for display.
4. The output includes the temperature unit.

- **Use cases:** Simulating sensor readings, testing price calculations, generating sample measurements, and Monte Carlo models.
- **Note:** uniform() generates floating-point values; its exact upper-bound behavior can depend on floating-point rounding.

# Example 5: Randomly select an item using choice()
* Problem statement
* You have a list of machine learning algorithms and want to randomly select one for a demonstration.

```python
import random

algorithms = [
    "Linear Regression",
    "Logistic Regression",
    "Decision Tree",
    "Random Forest",
    "KNN"
]

selected_algorithm = random.choice(algorithms)

print("Selected algorithm:", selected_algorithm)
```

## Step-by-step explanation
1. Create a list of algorithm names.
2. random.choice(algorithms) selects one element from the list.
3. Store the selected element in selected_algorithm.
4. Print the result.

## You can also choose from strings, tuples, or other non-empty sequences.
- **Use cases:** Random quiz questions, game options, random test cases, and selecting a sample item from a list.
- **Important:** choice() selects one element. If the sequence is empty, it raises an IndexError.

# Example 6: Select multiple unique values using sample()
* Problem statement
* You have 20 student IDs and need to randomly select five different students for a survey, without selecting the same student twice.

```python
import random

student_ids = list(range(101, 121))

selected_students = random.sample(student_ids, 5)

print("Selected student IDs:", selected_students)
print("Number selected:", len(selected_students))
```

## Step-by-step explanation
1. range(101, 121) represents IDs from 101 through 120.
2. list() converts the range into a list.
3. random.sample(student_ids, 5) selects five distinct positions without replacement.
4. len() confirms that five IDs were selected.
* Use cases: Survey sampling, selecting test records, creating training samples, and splitting data for manual review.
* Important: sample() requires at least as many population elements as the requested sample size.

# 7. Bonus: Shuffle a list randomly

* Sometimes you want to change the order of all items rather than select only a few.

* Problem statement: Randomize the order of student names before conducting a quiz.

```python
import random

students = ["Amit", "Neha", "Rahul", "Priya", "Suraj"] # shuffle() modifies the original list in place and returns None.

random.shuffle(students)

print(students)
```

```python
# If you want to preserve the original list, copy it first:

students = ["Amit", "Neha", "Rahul", "Priya", "Suraj"]

shuffled_students = students.copy()
random.shuffle(shuffled_students)

print("Original:", students)
print("Shuffled:", shuffled_students)
```

## 8. Bonus: Generate repeatable random numbers with seed()
* **Problem statement:** You are testing a machine learning program and want to reproduce the same random sequence when you run it again.

```python
random.seed(42)

print(random.randint(1, 100))
print(random.randint(1, 100))
print(random.randint(1, 100))
```

### Step-by-step explanation
1. random.seed(42) initializes the pseudo-random generator with a fixed seed.
2. The first call generates a repeatable value.
3. Subsequent calls generate values from the same repeatable sequence.
4. Running the same code with the same Python random implementation and seed reproduces that sequence.


* A fixed seed is useful for debugging, testing, and reproducible experiments. In machine learning, you may also set random_state=42 in scikit-learn functions that support it.
