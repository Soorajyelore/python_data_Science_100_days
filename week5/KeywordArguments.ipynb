# Keyword Arguments

1. What Are Keyword Arguments in Python?

Keyword arguments are arguments passed to a function by specifying the parameter name and its value.

They allow you to pass values to a function using the parameter names, making your code easier to read and understand.

# Example 1: Without Keyword Arguments

```python
def student(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)

student("Suraj", 30, "Pune")
```

Here, Python assigns the values based on their positions:

- "Suraj" → name
- 30 → age
- "Pune" → city

This is an example of positional arguments.

# Example 2: With Keyword Arguments

```python
def student(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)

student(name="Suraj", age=30, city="Pune")
```

#### Step-by-step explanation
- Step 1: Define the function.

- def student(name, age, city):


* We create a function named student with three parameters.

#### Step 2: Pass keyword arguments.

- student(name="Suraj", age=30, city="Pune")

* We explicitly specify which value belongs to which parameter.

#### Step 3: Python matches the names.

*Python assigns:*
- name = "Suraj"
- age = 30
- city = "Pune"


**Important advantage: You can change the order of keyword arguments without changing the result.**

**student(city="Pune", name="Suraj", age=30)**

*This works because Python matches values by parameter name, not by position.*

# Example 3: Keyword Arguments with Default Parameters
* Problem statement


Create a function that calculates the final price of a product after applying a discount.

If the customer does not specify a discount, use a default discount of 10%.
Python code

```python
def calculate_discount(price, discount=10):
    discount_amount = price * discount / 100

    final_price = price - discount_amount

    print("Original Price:", price)
    print("Discount:", discount, "%")
    print("Discount Amount:", discount_amount)
    print("Final Price:", final_price)


# Calling the function without specifying the discount
calculate_discount(price=1000)

print()
```

* Step 1: Define a function with a default value.
- def calculate_discount(price, discount=10):

Here:
- price is a required parameter.
- discount=10 means the default discount is 10%.


Step 2: Calculate the discount amount.

discount_amount = price * discount / 100

If the price is ₹1,000 and the discount is 10%:

1000 × 10 / 100 = 100

Step 3: Calculate the final price.

final_price = price - discount_amount

The final price is ₹1,000 − ₹100 = ₹900.

Step 4: Call the function without a discount.

calculate_discount(price=1000)

Python automatically uses the default discount of 10%.

Step 5: Override the default discount.

calculate_discount(price=1000, discount=20)

Python uses the supplied discount of 20% instead.

Real-world use case: E-commerce websites, promotional offers, and billing systems.

Important: Default parameter values are used only when you do not provide a value for that parameter.

### Example 4: Mixing Positional and Keyword Arguments
### Problem statement

Create a function that calculates a student's percentage based on marks obtained in three subjects.

Pass the student's name as a positional argument and the marks as keyword arguments.
Python code

```python
def calculate_percentage(name, maths, science, english):
    total = maths + science + english

    percentage = (total / 300) * 100

    print("Student Name:", name)
    print("Total Marks:", total)
    print("Percentage:", percentage, "%")


calculate_percentage(
    "Suraj",#positional argument, while the subject marks are keyword arguments.
    maths=85,
    science=90,
    english=80
)
```

## Example 5: Keyword Arguments with **kwargs
# Problem statement
Create a function that accepts an employee's name and any number of additional details, such as department, city, salary, and experience.

The function should display all the supplied details

```python
def employee_details(name, **kwargs):# name is a regular parameter.
    print("Employee Name:", name)

    print("\nAdditional Details:")

    for key, value in kwargs.items():# **kwargs collects additional keyword arguments into a dictionary
        print(key, ":", value)


employee_details(
    "Suraj",
    department="Data Science",
    city="Pune",
    salary=60000,
    experience=5
)
```

The function has two parts:
- name is a regular parameter.
- **kwargs collects additional keyword arguments into a dictionary.

For this function call, Python collects the additional details into a dictionary similar to this:

## Step 3: Access the additional details.

for key, value in kwargs.items():


## The .items() method returns each dictionary key and its corresponding value.

For example:
- key = "department" and value = "Data Science"
- key = "city" and value = "Pune"
- key = "salary" and value = 60000

# Step 4: Print each detail.

print(key, ":", value)


The loop prints every additional parameter.


## Try adding more keyword arguments

-The same function can handle the additional details without requiring changes to its definition.

-Real-world use case: Flexible configuration functions, API parameters, employee information systems, and reusable utility functions.

* Remember: **kwargs is a Python convention. You could use another parameter name, but the double asterisks are what collect the keyword arguments into a dictionary.

## 4. Important Rules to Remember
* Rule 1: Keyword arguments use parameter names


Correct:

```python
def add(a, b):
    print(a + b)

add(a=10, b=20)
```

Rule 2: Do not provide the same argument twice

Incorrect:

```python
def student(name, age):
    print(name, age)

student("Suraj", name="Amit", age=30)
```

Rule 3: Required parameters must receive values

```python
def student(name, age):
    print(name, age)

student(name="Suraj")
```

Rule 4: Keyword arguments can override defaults

```python
def greet(name, message="Hello"):
    print(message, name)

greet(name="Suraj")
greet(name="Suraj", message="Welcome")
```

Rule 5: **kwargs collects extra keyword arguments

```python
def show_details(**kwargs):
    print(kwargs)

show_details(name="Suraj", age=30, city="Pune")
```

Final Summary
The most important concepts to remember are:
- Keyword arguments: Pass values using parameter names, such as name="Suraj".
- Default arguments: Provide a value that Python uses when an argument is omitted.
- Mixed arguments: Use positional arguments before keyword arguments.
- **kwargs: Collect any number of additional keyword arguments into a dictionary.
- Practical applications: Keyword arguments make Python functions easier to read, configure, reuse, and maintain.
