# break, continue, and pass in Python

These three statements are used to control the execution of loops and code blocks.

| Statement  | Meaning    | What it does                                        |
| ---------- | ---------- | --------------------------------------------------- |
| `break`    | Stop       | Completely stops the loop                           |
| `continue` | Skip       | Skips current iteration and moves to next iteration |
| `pass`     | Do nothing | Placeholder; does nothing when executed             |


# A simple way to remember:

* break    → STOP the loop
* continue → SKIP this iteration
* pass     → DO NOTHING

# 1. break Statement

The break statement immediately terminates the loop.

### Example 1: Stop when number is 5

```python
for i in range(1, 11):

    if i == 5:
        break

    print(i)
```

# 2. break with a List

Suppose we are searching for a particular name.

```python
names = ["John", "Sam", "David", "Rahul", "Alex"]

for name in names:

    print("Checking:", name)

    if name == "Rahul":
        print("Found Rahul")
        break
```

# Note:- 

Real-world use

This is useful when searching for something and you don't need to continue once you've found it.

# 3. continue Statement

The continue statement skips the current iteration and moves to the next iteration.

##### Example 3: Skip number 5

```python
for i in range(1, 11):

    if i == 5:
        continue

    print(i)
```

Notice that 5 is missing.

# 4. continue with Even/Odd Numbers

Suppose we only want to print odd numbers.

```python
for i in range(1, 11):

    if i % 2 == 0:
        continue

    print(i)
```

# 5. pass Statement

* pass means:

Do nothing.

It is mainly used as a placeholder when you want to write a block of code but don't want to implement it yet.

### Example 5

```python
for i in range(1, 6):

    if i == 3:
        pass

    print(i)
```

# 6. pass in a Function

pass is also useful when creating a function that you plan to implement later.

```python
def calculate_salary():
    pass
```

Python accepts this code.

Without pass, this would cause an error:

Later, you can replace pass:

```python
calculate_salary()
```

So pass is essentially saying:

🔥 Difference Between break, continue, and pass

Look at this example:

# break


```python
for i in range(1, 6):

    if i == 3:  # 3 stops the entire loop.
        break

    print(i)
```

# continue

```python
for i in range(1, 6):

    if i == 3:    # 3 skips the current iteration.
        continue

    print(i)
```

# pass

```python
for i in range(1, 6):

    if i == 3: # 3 does nothing, but the loop continues.
        pass

    print(i)
```

| Can be used in functions?   | ❌            | ❌            | ✅      |

# 8. Practical Example: Processing Student Marks

Suppose we have student marks:

```python
marks = [45, 78, -1, 65, 90, 32]
```

-1 represents invalid data.

We can use continue to skip it.

```python
marks = [45, 78, -1, 65, 90, 32]

for mark in marks:

    if mark == -1:
        continue

    print("Valid Mark:", mark)
```

# 9. Practical Example: Stop at Invalid Data

Now suppose we want to stop processing as soon as we find an invalid value.

```python
marks = [45, 78, 65, -1, 90, 32]

for mark in marks:

    if mark == -1:
        print("Invalid data found!")
        break

    print("Processing:", mark)
```

Here:

break

stops the entire loop.

# 🧠 Very Important: Nested Loops

Since you just learned nested loops, understand how break behaves there.

```python
for i in range(1, 4):

    for j in range(1, 4):

        if j == 2:
            break

        print(i, j)
```

# 🎯 Interview Questions
#### 1. What is break?

* break is used to immediately terminate a loop and transfer control to the statement after the loop.

#### 2. What is continue?

* continue skips the current iteration and moves to the next iteration of the loop.

#### 3. What is pass?

pass is a null statement that performs no operation. It is commonly used as a placeholder when a block of code is required syntactically but implementation is not ready.

#### 4. Difference between break and continue?

* break terminates the entire loop, while continue skips only the current iteration and allows the loop to continue.

#### 5. Does pass skip an iteration?

* No.

* break    → STOP
* continue → SKIP
* pass     → DO NOTHING
