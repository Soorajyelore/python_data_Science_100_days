# 🔄 For Loop in Python

A for loop is used to repeat a block of code for every item in a sequence such as a list, tuple, string, dictionary, or range.

# Basic Syntax

```python
for i in range(1, 6): #The loop takes each value from range(1, 6) one by one and stores it in i.
    print(i)
```

# Example 1: Print Numbers 1 to 5


# step by step 



| Iteration | `i` | Output |
| --------- | --: | -----: |
| 1         |   1 |      1 |
| 2         |   2 |      2 |
| 3         |   3 |      3 |
| 4         |   4 |      4 |
| 5         |   5 |      5 |


# Important
* range(1, 6)

# means:

* 1, 2, 3, 4, 5

# The ending value 6 is not included.

# Example 2: Print Even Numbers

```python
for i in range(2, 11, 2):
    print(i)
```

# Step-by-step

### The syntax is:

* range(start, stop, step)

### So:

* range(2, 11, 2)

### means:

* Start = 2
* Stop before = 11
* Increase by = 2

### Therefore:

* 2 → 4 → 6 → 8 → 10

# Example 3: Loop Through a List

* A for loop can directly access each item from a list.

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

for fruit in fruits:
    print(fruit)
```

# Example 4: Calculate Sum of Numbers

* Suppose we want to calculate:

* 10 + 20 + 30 + 40 + 50

# We can use a for loop.

```python
numbers = [10, 20, 30, 40, 50]

total = 0

for num in numbers:
    total = total + num

print("Total:", total)
```

# Example 5: Check Even or Odd

* We can combine a for loop with an if statement.

```python
numbers = [10, 15, 22, 31, 40]

for num in numbers:

    if num % 2 == 0:
        print(num, "is Even")
    else:
        print(num, "is Odd")
```

# Example 6: Loop Through a String

* A string is also a sequence, so we can use a for loop to access each character.

```python
name = "Python"

for char in name:
    print(char)
```

# Example 7: Multiplication Table

* A very useful beginner example is creating a multiplication table.

```python
num = 5

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

# ⭐ for Loop with range()

# for Loop vs while Loop

| `for` loop                            | `while` loop                                 |
| ------------------------------------- | -------------------------------------------- |
| Used when iterating over a sequence   | Used when condition controls repetition      |
| Common with lists, strings, `range()` | Common when number of iterations isn't known |
| Usually simpler                       | More flexible for condition-based loops      |
| Automatically moves to next item      | You usually need to update the variable      |
| Example: `for x in list`              | Example: `while x < 10`                      |


```python
# Example

# For loop:

for i in range(5):
    print(i)

# While loop:

i = 0

while i < 5:
    print(i)
    i += 1
```
