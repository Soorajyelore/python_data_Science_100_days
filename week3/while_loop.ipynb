# While Loop in Python

A while loop is used when you want to execute a block of code repeatedly as long as a condition is True.

It is especially useful when you don't know in advance how many times the loop should run.

### Basic Syntax

# 1. Basic While Loop

* Let's print numbers from 1 to 5.

```python
i = 1 # Initially:

while i <= 5: # Iteration 1
    print(i)
    i = i + 1
```

# 2. Print Even Numbers

* We can use a while loop to print even numbers from 2 to 10.

```python
i = 2

while i <= 10:
    print(i)
    i = i + 2
```

# Step-by-step

i = 2
2 <= 10 → True → print 2
i = 4

4 <= 10 → True → print 4
i = 6

6 <= 10 → True → print 6
i = 8

8 <= 10 → True → print 8
i = 10

10 <= 10 → True → print 10
i = 12

12 <= 10 → False → Stop

Here we increase i by 2 each time.

# 3. Calculate Sum of Numbers

* Let's calculate the sum of numbers from 1 to 5.

```python
i = 1
total = 0

while i <= 5:
    total = total + i
    i = i + 1

print("Total:", total)
```

# Step-by-step

# 4. While Loop with User Input

* A very common use of while is repeatedly asking the user for input.

* For example, keep asking for a password until the correct password is entered.

```python
password = ""

while password != "python123":
    password = input("Enter password: ")

print("Login successful!")
```

# 5. While Loop with break

* The break statement is used to immediately stop a loop.

* Example: Stop When Number Reaches 5

```python
i = 1

while True:
    print(i)

    if i == 5:
        break

    i = i + 1
```

# 6. While Loop with continue

* The continue statement skips the current iteration and moves to the next iteration.

####  Example: Print Only Odd Numbers

```python
i = 0

while i < 10:
    i = i + 1

    if i % 2 == 0:
        continue

    print(i)
```

# Step-by-step

# 7. Practical Example: Countdown

* A countdown is another simple and useful example.

```python
count = 5

while count >= 1:
    print(count)
    count = count - 1

print("Happy New Year!")
```

# while vs for Loop

| `while` loop                                | `for` loop                               |
| ------------------------------------------- | ---------------------------------------- |
| Condition-based                             | Sequence/range-based                     |
| Useful when number of iterations is unknown | Useful when iterations are known         |
| Can continue until a condition changes      | Usually iterates over a collection/range |
| `while condition:`                          | `for item in sequence:`                  |


```python
# Example

# While:

i = 1

while i <= 5:
    print(i)
    i += 1

# For:

for i in range(1, 6):
    print(i)
```

# Important Interview Points

### What is a while loop?

* A while loop repeatedly executes a block of code as long as a specified condition is True.

### What is break?

* break immediately terminates the loop.

### What is continue?

* continue skips the current iteration and moves to the next iteration.

### What causes an infinite loop?

* An infinite loop occurs when the loop condition always remains True, often because the variable controlling the condition is not updated.
