# String Methods and String Indexing in Python

* Strings are one of the most commonly used data types in Python, especially in data cleaning, data analysis, and text processing.

# There are two important concepts:

* String Indexing → accessing characters using their position.
* String Methods → performing operations such as converting case, searching, replacing, splitting, etc.

# 1. String Indexing

* String indexing is used to access individual characters from a string.

* Python uses zero-based indexing, which means the first character has index 0.

```python
text = "Python"
```

# The indexes are:

# Types of indexing

| Type              | Syntax       | Purpose                        |
| ----------------- | ------------ | ------------------------------ |
| Positive indexing | `text[0]`    | Access from left               |
| Negative indexing | `text[-1]`   | Access from right              |
| Slicing           | `text[1:4]`  | Extract part of string         |
| Step slicing      | `text[::2]`  | Extract characters with a step |
| Reverse slicing   | `text[::-1]` | Reverse string                 |


# 2. String Methods

* String methods are built-in functions that operate on strings.

* Some important methods are:

| Method         | Purpose                          |
| -------------- | -------------------------------- |
| `upper()`      | Convert to uppercase             |
| `lower()`      | Convert to lowercase             |
| `title()`      | Convert to title case            |
| `strip()`      | Remove spaces from beginning/end |
| `replace()`    | Replace text                     |
| `split()`      | Convert string into a list       |
| `join()`       | Join list elements               |
| `find()`       | Find position of text            |
| `count()`      | Count occurrences                |
| `startswith()` | Check beginning                  |
| `endswith()`   | Check ending                     |
| `isdigit()`    | Check for digits                 |
| `isalpha()`    | Check for alphabets              |
| `isalnum()`    | Check for alphabets/numbers      |


# Example 1: Indexing + Case Methods

```python
text = "python programming"

# Indexing
print("First character:", text[0])
print("Last character:", text[-1])

# String methods
print("Upper:", text.upper())
print("Lower:", text.lower())
print("Title:", text.title())
```

# Example 2: String Slicing

* Slicing is used to extract a portion of a string.

# Syntax

#### string[start:stop]

* The start index is included, but stop is excluded.

```python
text = "Python Programming"

print(text[0:6])
print(text[7:18])
print(text[:6])
print(text[7:])
```

# Example with step

```python
text = "Python"

print(text[::2])
print(text[::-1])
```

# Example 3: strip(), replace() and title()

* These methods are very useful in data cleaning.

```python
name = "   suraj yelore   "

print("Original:", name)
print("Cleaned:", name.strip())
print("Title:", name.strip().title())

new_name = name.strip().replace("yelore", "Yelore")

print("After replace:", new_name)
```

# Example 4: split(), join() and Indexing

```python
skills = "Python,SQL,Machine Learning,Statistics"

# Convert string into list
skill_list = skills.split(",")

print(skill_list)

# Indexing the list
print("First skill:", skill_list[0])
print("Last skill:", skill_list[-1])

# Convert list back into string
result = " | ".join(skill_list)

print(result)
```

# Example 5: Searching and Checking a String

### Methods used:

* find()
* count()
* startswith()
* endswith()

```python
text = "Python is easy to learn"

print("Position of 'easy':", text.find("easy"))
print("Count of 'Python':", text.count("Python"))

print("Starts with Python:", text.startswith("Python"))
print("Ends with learn:", text.endswith("learn"))
```

# Example 6: Checking String Content

* Methods such as isalpha(), isdigit(), and isalnum() are useful for validation.

```python
name = "Suraj"
age = "35"
employee_id = "EMP123"

print(name.isalpha())
print(age.isdigit())
print(employee_id.isalnum())
```

# Practical Data Analytics Example

* Suppose you receive customer names like this:

```python
names = [
    "  suraj ",
    "PRIYA",
    " rahul",
    "AMIT  "
]
```

# We want to clean the data.

```python
# This is a very common data-cleaning operation.

names = [
    "  suraj ",
    "PRIYA",
    " rahul",
    "AMIT  "
]

clean_names = []

for name in names:
    cleaned = name.strip().title()
    clean_names.append(cleaned)

print(clean_names)
```

| Negative indexing | `text[-1]`       | Get last character |
| Slicing           | `text[0:5]`      | Get part of string |

| `lower()`         | `text.lower()`   | Convert case       |

####  Important Interview Questions
* 1. What is string indexing?

####  String indexing is the process of accessing individual characters from a string using their position. Python starts indexing from 0.

* 2. What is negative indexing?

####  Negative indexing allows us to access characters from the end of a string. -1 represents the last character.

* 3. What is string slicing?

#####  String slicing is used to extract a portion of a string using the syntax string[start:stop:step].

* 4. Are Python strings mutable?

#####  No. Python strings are immutable, meaning we cannot change an existing string directly. String methods return a new string.
