# Dictionaries in Python – Types, Methods

# 1. What is a dictionary in Python?
A dictionary is a collection of data stored as key-value pairs. Each key is used to access its corresponding value.



###  For example, you can store an employee's name, salary, department, and experience in a dictionary.
####  Syntax:

```python
employee = {
    "name": "S",
    "salary": 55000,
    "department": "Data Science",
    "experience": 5
}

print(employee)
print(employee["name"])
print(employee["salary"])
```

Explanation:
- "name" is a key, and "Suraj" is its value.
- "salary" is a key, and 55000 is its value.
- You access a value using its key, such as employee["name"].
- Dictionaries are mutable, so you can add, update, and remove entries.
- Keys must be unique and hashable. Values can be of different data types.

# 3. Python code examples with explanations




#### Example 1: Access and update employee information
* Problem statement: Create an employee dictionary, retrieve employee details, and update the salary.

```python
employee = {
    "name": "Suraj",
    "salary": 50000,
    "department": "Data Science"
}

print("Employee name:", employee["name"])
print("Old salary:", employee["salary"])

employee["salary"] = 60000

print("Updated salary:", employee["salary"])
```

Explanation:
1. Create a dictionary with employee details.
2. Use employee["name"] to retrieve the name.
3. Use employee["salary"] to retrieve the salary.
4. Assign a new value to employee["salary"] to update it.



To add a new key, you can write:


employee["experience"] = 5


print(employee)


This adds the "experience" key with a value of 5.

# Example 2: Add, delete, and retrieve dictionary values
Problem statement: Manage a product's information in an inventory syste

```python
product = {
    "product_id": 101,
    "name": "Laptop",
    "price": 55000
}

# Add a new key
product["stock"] = 20

# Retrieve a value safely
print("Brand:", product.get("brand", "Not available"))

# Delete a key
del product["stock"]

print(product)
```

Explanation:
- product["stock"] = 20 adds a new key-value pair.
- get("brand", "Not available") returns a default value because "brand" is absent.
- del product["stock"] deletes the stock entry.



* Important difference:


print(product["brand"])  # Raises KeyError if missing


print(product.get("brand"))  # Returns None if missing



Use get() when a key may not exist and you want to handle it safely.

# Example 3: Loop through a dictionary
Problem statement: Print all employee details, one by one.

```python
employee = {
    "name": "Amit",
    "age": 28,
    "salary": 45000,
    "department": "IT"
}

for key, value in employee.items():
    print(key, ":", value)
```

Explanation:
- employee.items() provides each key-value pair.
- key stores the key.
- value stores the corresponding value.
- The for loop repeats for every pair.


Other useful looping methods:


print(employee.keys())    # All keys


print(employee.values())  # All values

These methods are helpful when you want to process only keys or only values.

# Example 4: Calculate total and average marks
Problem statement: Store student marks in a dictionary and calculate the total and average.

```python
marks = {
    "Maths": 80,
    "Science": 90,
    "English": 85,
    "Python": 95
}

total = sum(marks.values())
number_of_subjects = len(marks)
average = total / number_of_subjects

print("Total marks:", total)
print("Number of subjects:", number_of_subjects)
print("Average marks:", average)
```

Explanation:
1. marks.values() retrieves the marks.
2. sum() adds all the marks.
3. len(marks) counts the subjects.
4. Divide total marks by the number of subjects to calculate the average.


This is a simple example of using dictionaries for data analysis.

# Example 5: Nested dictionary — manage multiple employees
Problem statement: Store multiple employee records and print the details of a particular employee.

```python
employees = {
    "E101": {
        "name": "Amit",
        "salary": 40000
    },
    "E102": {
        "name": "Neha",
        "salary": 55000
    },
    "E103": {
        "name": "Rahul",
        "salary": 60000
    }
}

print(employees["E102"]["name"])
print(employees["E102"]["salary"])

for emp_id, details in employees.items():
    print(emp_id, details["name"], details["salary"])
```

Explanation:
- The outer dictionary uses employee IDs as keys.
- Each employee ID maps to another dictionary.
- employees["E102"]["name"] first retrieves employee E102, then accesses the "name" key.
- The loop processes each employee's record.


Nested dictionaries are useful for structured API responses, configuration files, and grouped datasets.

# Example 6: Count the frequency of each item
Problem statement: Count how many times each product appears in a sales list.

```python
products = [
    "Laptop",
    "Mouse",
    "Laptop",
    "Keyboard",
    "Mouse",
    "Laptop"
]

frequency = {}

for product in products:
    if product in frequency:
        frequency[product] += 1
    else:
        frequency[product] = 1

print(frequency)
```

Explanation:
1. Start with an empty dictionary.
2. Loop through each product.
3. If the product already exists as a key, increase its count by one.
4. Otherwise, create a key with an initial count of one.


This is a common Python interview question and a useful technique for counting categories in data analytics.
A shorter alternative uses collections.Counter:

```python
from collections import Counter

frequency = Counter(products)
print(frequency)
```

```python
employee = {"name": "Suraj", "salary": 50000}

employee.update({"salary": 60000, "city": "Pune"})
removed_salary = employee.pop("salary")

print(employee)
print("Removed salary:", removed_salary)
```

# 5. Common mistakes to avoid
Mistake 1: Accessing a missing key

```python
data = {"name": "Suraj"}

# print(data["age"])  # KeyError
print(data.get("age", 0))  # 0
```

# Mistake 2: Repeating a key

```python
data = {"name": "Amit", "name": "Rahul"}
print(data)
```

The later value replaces the earlier value.
# Mistake 3: Confusing a list with a dictionary

```python
# List: accessed by position
numbers = [10, 20, 30]
print(numbers[0])

# Dictionary: accessed by key
employee = {"id": 101}
print(employee["id"])
```
