Lists are especially important in Data Analytics and Data Science because they are commonly used to store numbers, names, records, features, and other collections of data.

# 4. List Indexing

List indexing starts from [0].

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[2])
print(numbers[4])
```

# Negative Indexing

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[-1])
print(numbers[-2])
```

# 5. List Slicing

You can extract part of a list using slicing.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
```

# 6. Adding Items to a List
append()

Adds one item to the end.

```python
numbers = [10, 20, 30]

numbers.append(40)

print(numbers)
```

# insert()

Adds an item at a specific position.

```python
numbers = [10, 20, 30]

numbers.insert(1, 15)

print(numbers)
```

# extend()

Adds multiple items.

```python
numbers = [10, 20, 30]

numbers.extend([40, 50, 60])

print(numbers)
```

# Difference

* append([40])

# adds the whole list as one element.

* extend([40, 50])

adds the elements individually.

# 7. Removing Items
remove()

Removes a specific value.

```python
numbers = [10, 20, 30, 40]

removed = numbers.pop(2)

print("Removed:", removed)
print(numbers)

```

If you use:

numbers.pop()

it removes the last element.

# del

You can also delete an item using del.

```python
numbers = [10, 20, 30, 40]

del numbers[1]

print(numbers)
```

# clear()

Removes everything.

```python
numbers = [10, 20, 30]

numbers.clear()

print(numbers)
```

# Example 1: Find Maximum and Minimum
Problem

Given a list of employee salaries, find the minimum and maximum salary.

```python
salaries = [25000, 45000, 32000, 75000, 50000]

maximum = max(salaries)
minimum = min(salaries)

print("Maximum Salary:", maximum)
print("Minimum Salary:", minimum)
```

# Example 2: Find Even and Odd Numbers
Problem

Given a list of numbers, separate even and odd numbers.

```python
numbers = [10, 15, 22, 31, 40, 55, 60]

even_numbers = []
odd_numbers = []

for num in numbers:

    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)

print("Even:", even_numbers)
print("Odd:", odd_numbers)
```

# Example 3: Find Duplicate Values
Problem

Find duplicate values from a list.

```python
numbers = [10, 20, 30, 20, 40, 10, 50]

duplicates = []

for num in numbers:

    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print("Duplicates:", duplicates)
```

# Example 4: Calculate Average
Problem

You have student marks. Calculate the average.

```python
marks = [70, 80, 65, 90, 75]

total = 0

for mark in marks:
    total = total + mark

average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

# Example 5: Filter Data Using Conditions
Problem

Suppose we have employee salaries. Find employees whose salary is greater than ₹50,000.

```python
salaries = [30000, 55000, 45000, 75000, 60000, 40000]

high_salary = []

for salary in salaries:

    if salary > 50000:
        high_salary.append(salary)

print("High Salaries:", high_salary)
```

# Example 6: Nested List / Matrix
Problem

Print every value from a 2D list.

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

for row in matrix:

    for value in row:
        print(value, end=" ")

    print()
```

# 9. Sorting a List
Ascending order

```python
numbers = [50, 10, 40, 20, 30]

numbers.sort()

print(numbers)
```

# Descending order 

```python
numbers = [50, 10, 40, 20, 30]

numbers.sort(reverse=True)

print(numbers)
```

# 10. Reverse a List

```python
numbers = [10, 20, 30, 40, 50]

numbers.reverse()

print(numbers)
```

# 11. Check Whether an Item Exists

Use the in operator.

```python
names = ["John", "Sam", "David", "Alex"]

if "David" in names:
    print("David is present")
else:
    print("David is not present")
```

```python
if "Rahul" not in names:
    print("Rahul is not present")
```

# 12. List Comprehension

List comprehension is a short way to create a list.

Instead of:

```python
numbers = [1, 2, 3, 4, 5]

squares = []

for num in numbers:
    squares.append(num * num)

print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6]

even = [num for num in numbers if num % 2 == 0]

print(even)
```

# ⚠️ Important: Lists Are Mutable

Mutable means you can change the list after creating it.

```python
numbers = [10, 20, 30]

numbers[1] = 100

print(numbers)
```
