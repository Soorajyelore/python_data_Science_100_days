# Tuples in Python 
A tuple in Python is a collection used to store multiple values in a single variable. A tuple is similar to a list, but the main difference is that a tuple is immutable, meaning you cannot change its elements after creating it.


Tuples are useful for storing fixed data, such as employee details, coordinates, dates, and database records.

# 1. Creating a tuple in Python
You can create a tuple using parentheses ().

```python
numbers = (10, 20, 30, 40, 50)

print(numbers)
print(type(numbers))
```

# 2. Different types of tuples

# Example 1: Tuple indexing and slicing
Problem: Store employee details and retrieve particular elements.

```python
employee = ("Suraj", 55000, "Data Scientist", 5)

print(employee[0])
print(employee[2])
print(employee[-1])
print(employee[1:3])
```

#### Step-by-step explanation
The tuple contains four elements:

* employee[0] returns the first element.
* employee[2] returns the third element.
* employee[-1] returns the last element.
* employee[1:3] returns elements at indexes 1 and 2. The stop index 3 is excluded.
Remember: Tuple indexing starts at 0, just like list indexing.

# Example 2: Loop through a tuple
Problem: Print each skill stored in a tuple.

```python
skills = ("Python", "SQL", "Machine Learning", "Pandas")

for skill in skills:
    print(skill)
```

Step-by-step explanation
1. Python takes "Python" and stores it in skill.
2. It executes print(skill).
3. It moves to "SQL" and prints it.
4. It repeats for "Machine Learning" and "Pandas".
5. The loop ends when all elements have been processed.
You can also use a condition inside the loop:

```python
marks = (45, 78, 32, 90, 65)

for mark in marks:
    if mark >= 60:
        print(mark, "Pass")
```

This demonstrates how tuples can be used with loops and conditional statements.

# Example 3: Tuple methods — count() and index()
Python tuples have two main built-in methods.
- count(value) counts how many times a value appears.
- index(value) returns the index of the first occurrence.

```python
numbers = (10, 20, 30, 20, 40, 20)

print(numbers.count(20))
print(numbers.index(30))
print(numbers.index(20))
```

Step-by-step explanation
* First statement:

```python
numbers.count(20)
```

The value 20 appears three times, so the result is 3.
* Second statement:

```python
numbers.index(30)
```

The value 30 is at index 2, so the result is 2.
* Third statement:

```python
numbers.index(20)
```

The first 20 is at index 1. Therefore, the result is 1, not 3.


Important: index() raises a ValueError if the requested value is not present in the tuple.

# Example 4: Why are tuples immutable?

Problem: Try to change an element of a tuple.

```python
numbers = (10, 20, 30)

numbers[1] = 100

print(numbers)
```

Result: Python raises a TypeError because tuple elements cannot be reassigned.

The correct approach, if you need changed values, is to create a new tuple:

```python
numbers = (10, 20, 30)

new_numbers = (numbers[0], 100, numbers[2])

print(new_numbers)
```

Step-by-step explanation
1. numbers contains (10, 20, 30).
2. We do not modify the original tuple.
3. We construct a new tuple, replacing the second value with 100.
4. new_numbers contains (10, 100, 30).


Interview point: A tuple is immutable, but if it contains a mutable object such as a list, that inner object may still be modified.
For example:

```python
data = (10, [20, 30], 40)

data[1].append(50)

print(data)
```

The tuple's references remain unchanged, but the list inside it changes.

# Example 5: Tuple unpacking
Problem: Store and retrieve employee information using tuple unpacking.

```python
employee = ("Suraj", 55000, "Data Scientist")

name, salary, role = employee

print("Name:", name)
print("Salary:", salary)
print("Role:", role)
```

Step-by-step explanation

The tuple contains three elements:

```python
employee = ("Suraj", 55000, "Data Scientist")
```

The number of variables must match the number of elements when using ordinary unpacking.

Tuple unpacking is useful for returning multiple values from functions and working with records.

# Example 6: Convert a list into a tuple
Problem: Convert a list of marks into a tuple and calculate the average.

```python
marks_list = [70, 80, 90, 60, 100]

marks_tuple = tuple(marks_list)

average = sum(marks_tuple) / len(marks_tuple)

print("Tuple:", marks_tuple)
print("Total:", sum(marks_tuple))
print("Average:", average)
```

Step-by-step explanation
1. marks_list stores five marks.
2. tuple(marks_list) converts the list into a tuple.
3. sum(marks_tuple) calculates the total: 400.
4. len(marks_tuple) returns the number of marks: 5.
5. The average is 400 / 5 = 80.0.

* Note that converting a list to a tuple does not automatically make the data better; it is useful when you want a fixed collection rather than a collection that you intend to modify.

```python
my_tuple = (10, 20, 30)
my_list = [10, 20, 30]

my_list.append(40)

print(my_list)
print(my_tuple)
```

# Interview answer: What is a tuple in Python?
A tuple is an ordered, immutable collection in Python that can store multiple values, including duplicate values and different data types. It supports indexing, slicing, iteration, and unpacking. Unlike lists, tuple elements cannot be reassigned, so tuples are useful for fixed data such as coordinates, configuration values, and records.
