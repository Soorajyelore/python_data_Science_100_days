# Sets in Python – Definition, Methods, and 6 Code Examples
#### 1. What is a Set in Python?
A set is a collection of unique elements. It automatically removes duplicate values.
Sets are useful in Python when you want to remove duplicates, find common values, or compare two collections of data.



Syntax:

```python
my_set = {10, 20, 30, 40}
print(my_set)
#checking the type of the variable
```

# 2. Python code examples


### Example 1: Remove duplicate values


* Problem: You have a list of employee IDs containing duplicates. Find the unique IDs.

```python
employee_ids = [101, 102, 103, 101, 104, 102, 105]

unique_ids = set(employee_ids)

print(unique_ids)
print("Total unique employees:", len(unique_ids))
```

Explanation:
- set(employee_ids) converts the list into a set.
- Duplicate IDs are automatically removed.
- len() counts the unique elements.

## Example 2: Add and remove elements


Problem: Add a new skill to an employee's skills set and remove a skill they no longer use.

```python
skills = {"Python", "SQL", "Excel"}

skills.add("Machine Learning")
print("After adding:", skills)

skills.remove("Excel")
print("After removing:", skills)

skills.discard("Java")
print("After discard:", skills)
```

Explanation:
- add() adds one element.
- remove() deletes an element but raises a KeyError if it doesn't exist.
- discard() deletes an element if present and does nothing if it is absent.

# Example 3: Union — combine two sets
Problem: Find all unique skills across two employees.

```python
employee1 = {"Python", "SQL", "Excel"}
employee2 = {"Python", "Power BI", "Tableau"}

all_skills = employee1.union(employee2)

print(all_skills)
```

# Example 4: Intersection — find common elements
Problem: Find the skills shared by two employees.

```python
employee1 = {"Python", "SQL", "Excel"}
employee2 = {"Python", "Power BI", "SQL"}

common_skills = employee1.intersection(employee2)

print(common_skills)
```

Explanation: Intersection returns only elements present in both sets.

# Example 5: Difference — find elements in one set but not another
* Problem: Find the skills Employee 1 has that Employee 2 does not have.

```python
employee1 = {"Python", "SQL", "Excel"}
employee2 = {"Python", "Power BI", "SQL"}

unique_to_employee1 = employee1.difference(employee2)

print(unique_to_employee1)
```

* Explanation: Difference returns elements from the first set that are absent from the second.

 The reverse difference, employee2 - employee1, returns {'Power BI'}.

```python
employee1 = {"Python", "SQL", "Excel"}
employee2 = {"Python", "Power BI", "SQL"}

unique_to_employee1 = employee2.difference(employee1)

print(unique_to_employee1)
```

# Example 6: Find common customers and unique customers
Problem: Two marketing campaigns reached different customer IDs. Find the common customers and customers reached by only one campaign.

```python
campaign_a = {101, 102, 103, 104, 105}
campaign_b = {103, 104, 105, 106, 107}

common_customers = campaign_a & campaign_b
only_campaign_a = campaign_a - campaign_b
only_campaign_b = campaign_b - campaign_a
all_customers = campaign_a | campaign_b

print("Common customers:", common_customers)
print("Only Campaign A:", only_campaign_a)
print("Only Campaign B:", only_campaign_b)
print("Total unique customers:", len(all_customers))
```

* Explanation: This example combines intersection, difference, union, and len(). These operations are useful in customer analytics, data cleaning, and comparing datasets.
