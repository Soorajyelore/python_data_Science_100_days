2D Lists in Python

A 2D list is a list that contains other lists.

You can think of it like a table or matrix:

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
```

Here:

* 3 rows
* 3 columns
* Total 9 values

# 1. Access Elements from a 2D List

The basic syntax is: 

# ist[row][column]

# Example


```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print(matrix[0][0])
print(matrix[1][2])
print(matrix[2][1])
```

# 2. Loop Through a 2D List

The most common way to process a 2D list is with nested for loops.

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
```

# 3. Student Marks Example

A 2D list can represent marks of multiple students.

```python
marks = [
    [80, 75, 90],
    [65, 70, 85],
    [90, 88, 95]
]

for student in marks:

    total = 0

    for mark in student:
        total = total + mark

    average = total / len(student)

    print("Total:", total)
    print("Average:", average)
    print()
```

# 4. Find the Largest Value in a 2D List
Problem

Find the largest number from the entire matrix.

```python
numbers = [
    [10, 25, 30],
    [45, 15, 60],
    [20, 80, 35]
]

largest = numbers[0][0]

for row in numbers:
    for num in row:

        if num > largest:
            largest = num

print("Largest:", largest)
```

# 5. Add Two Matrices

This is a very important 2D-list problem.

Suppose:

```python
matrix1 = [
    [1, 2],
    [3, 4]
]

matrix2 = [
    [5, 6],
    [7, 8]
]
```

```python
matrix1 = [
    [1, 2],
    [3, 4]
]

matrix2 = [
    [5, 6],
    [7, 8]
]

result = []

for i in range(len(matrix1)):

    row = []

    for j in range(len(matrix1[i])):
        value = matrix1[i][j] + matrix2[i][j]
        row.append(value)

    result.append(row)

print(result)
```

# 6. Find Even Numbers in a 2D List
Problem

Given a 2D list, print only the even numbers.

```python
numbers = [
    [10, 15, 20],
    [25, 30, 35],
    [40, 45, 50]
]

for row in numbers:

    for num in row:

        if num % 2 == 0:
            print(num)
```

# 7. 2D List with Employee Data

A practical Data Analytics example:

```python
employees = [
    ["John", 50000, 5],
    ["Sam", 65000, 7],
    ["David", 45000, 3],
    ["Alex", 80000, 10]
]

for employee in employees:
    name = employee[0]
    salary = employee[1]
    experience = employee[2]

    print(
        name,
        "Salary:", salary,
        "Experience:", experience
    )
```

```python
for employee in employees:

    if employee[1] > 60000:
        print(employee[0])
```
