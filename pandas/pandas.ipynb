# Example 2: Select columns, filter rows, and sort data
* Use case: An e-commerce company wants to find customers whose sales exceed ₹6,000 and arrange them from highest to lowest sales.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["Amit", "Priya", "Rahul", "Neha", "Sara"],
    "Age": [25, 30, 28, 35, 32],
    "City": ["Pune", "Mumbai", "Delhi", "Pune", "Mumbai"],
    "Sales": [5000, 8000, 6500, 9000, 7200]
})

# 1. Select one column
print(df["Customer"])

# 2. Select multiple columns
print(df[["Customer", "Sales"]])

# 3. Filter rows where sales are greater than 6000
high_sales = df[df["Sales"] > 6000]
print(high_sales)

# 4. Filter using two conditions
pune_customers = df[
    (df["City"] == "Pune") & (df["Sales"] > 6000)
]
print(pune_customers)

# 5. Sort by sales in descending order
sorted_df = df.sort_values("Sales", ascending=False)
print(sorted_df)
```

# 4. Example 2 — Select, filter, and sort data
Data selection
* Problem statement: From an employee dataset, find employees who work in IT and earn more than ₹40,000. Then sort them by salary.

```python

import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Vijay"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Experience": [3, 5, 2, 7, 4],
    "Salary": [45000, 60000, 35000, 80000, 50000]
}

df = pd.DataFrame(data)

# Step 1: Select a single column
print(df["Name"])

# Step 2: Select multiple columns
print(df[["Name", "Salary"]])

# Step 3: Filter employees working in IT
it_employees = df[df["Department"] == "IT"]
print(it_employees)

# Step 4: Filter using two conditions
result = df[
    (df["Department"] == "IT") &
    (df["Salary"] > 40000)
]
print(result)

# Step 5: Sort by salary from highest to lowest
sorted_df = df.sort_values(
    by="Salary",
    ascending=False
)
print(sorted_df)

# Step 6: Select rows by position
print(df.iloc[0])

# Step 7: Select using row and column labels
print(df.loc[0, "Name"])
```

## 5. Example 3 — Clean missing values, duplicates, and incorrect data
* Data cleaning
* Problem statement: A customer dataset contains missing ages, missing salaries, duplicate records, and an invalid negative age. Clean it before analysis.
#### Python code

```python

import pandas as pd
import numpy as np

# Step 1: Create a messy dataset
data = {
    "Name": ["Amit", "Priya", "Rahul", "Priya", "Sneha"],
    "Age": [25, np.nan, 30, np.nan, -5],
    "Salary": [40000, 60000, np.nan, 60000, 50000]
}

df = pd.DataFrame(data)

print("Original data:")
print(df)

# Step 2: Check missing values
print(df.isnull().sum())

# Step 3: Remove duplicate rows
df = df.drop_duplicates()

# Step 4: Replace invalid ages with missing values
df.loc[df["Age"] < 0, "Age"] = np.nan

# Step 5: Fill missing ages with median age
age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)

# Step 6: Fill missing salaries with median salary
salary_median = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(salary_median)

# Step 7: Check the cleaned data
print("Cleaned data:")
print(df)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

* What happens at every step?
1. Create the dataset. np.nan represents a missing numeric value. We use NumPy here because it provides a common missing-value marker for numeric data.
2. Check missing values. df.isnull() returns True wherever a value is missing. Adding .sum() counts the missing values in each column.
3. Remove duplicates. drop_duplicates() removes rows that are identical across all columns. Here, the repeated Priya record is removed.
4. Fix invalid values. Negative age is not valid in this example, so we replace it with NaN rather than treating it as a real age.
5. Calculate the median. median() finds the middle value of the available numbers. The median is often less sensitive to extreme values than the mean.
6. Fill missing values. fillna() replaces missing values with the chosen median.
7. Validate the result. The final isnull().sum() confirms that no missing values remain in these two columns.

# Other useful cleaning functions
|Function	|Purpose
:---------|----------
df.dropna()	|Remove rows containing missing values by default.
df.fillna(0)	|Replace missing values with zero. Use only when zero is meaningful.
df.replace("?", np.nan)	|Replace a placeholder with a missing value.
df.drop_duplicates()	|Remove duplicate rows.
df.rename(columns={"Name": "Customer"})	|Rename a column.
df.astype({"Age": "int64"})	|Convert a column to a specified type, if the values permit it.

**Important:** In real projects, don't automatically replace every missing value with zero or the median. Choose a method based on what the missing value means, the data distribution, and the business problem. For predictive modeling, calculate imputation values using the training data only to avoid data leakage.


**Real-world use case:** Cleaning customer records, insurance claims, sales data, or healthcare data before exploratory data analysis (EDA) and model training.

# 6. Example 4 — GroupBy, aggregation, and business analysis


**Data analysis**
This is one of the most useful Pandas concepts for Data Analyst and Data Scientist interviews.
- Problem statement: A company wants to know the average salary by department, total salary expenditure by department, and the number of employees in each department.

```python

import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Vijay", "Neha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [45000, 60000, 35000, 80000, 50000, 55000],
    "Experience": [3, 5, 2, 7, 4, 4]
}

df = pd.DataFrame(data)

# Step 1: Group employees by department
grouped = df.groupby("Department")

# Step 2: Calculate average salary
print(grouped["Salary"].mean())

# Step 3: Calculate total salary
print(grouped["Salary"].sum())

# Step 4: Count employees
print(grouped["Name"].count())

# Step 5: Calculate multiple metrics together
summary = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Total_Salary=("Salary", "sum"),
    Employee_Count=("Name", "count"),
    Max_Experience=("Experience", "max")
)

print(summary)

# Step 6: Sort departments by average salary
print(summary.sort_values(
    by="Average_Salary",
    ascending=False
))
```

# 7. Example 5 — Merge and concatenate datasets
* Combining data
In real projects, related information is often stored in different files or database tables. Pandas lets you combine them.


**Problem statement:** You have one dataset containing customer details and another containing customer purchases. Combine them to find each customer's name, city, and purchase amount.

```python

import pandas as pd

# Dataset 1: Customer details
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Name": ["Amit", "Priya", "Rahul", "Sneha"],
    "City": ["Pune", "Mumbai", "Nashik", "Pune"]
})

# Dataset 2: Purchase details
orders = pd.DataFrame({
    "Customer_ID": [101, 102, 102, 105],
    "Amount": [5000, 8000, 3000, 6000]
})

print("Customers:")
print(customers)

print("Orders:")
print(orders)

# Step 1: Inner join
inner_result = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="inner"
)
print(inner_result)

# Step 2: Left join
left_result = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="left"
)
print(left_result)

# Step 3: Outer join
outer_result = pd.merge(
    customers, orders,
    on="Customer_ID",
    how="outer"
)
print(outer_result)

# Step 4: Concatenate datasets vertically
extra_customers = pd.DataFrame({
    "Customer_ID": [106],
    "Name": ["Karan"],
    "City": ["Nagpur"]
})

all_customers = pd.concat(
    [customers, extra_customers],
    ignore_index=True
)

print(all_customers)
```

## Understand the difference
* 1. merge() — match related records
Joins datasets using a common key, such as Customer_ID.
Use it when customer details and purchase details are stored separately.


* 2. Inner join

Keeps only IDs found in both datasets. Here, IDs 101 and 102 match, and ID 102 appears twice because there are two orders.


* 3. Left join

Keeps every customer from the left dataset, including customers without an order. Sneha's order amount will be missing.


* 4. Outer join

Keeps all IDs from both datasets. This includes customer ID 105, which appears in the orders data but not in the customer data.


* 5. concat() — stack data

Adds the new customer row below the existing rows. ignore_index=True creates a fresh sequential index.

## Join types at a glance
|Join|	Records retained
:-----|-------------------
|inner|	Matching keys from both datasets
|left|	All left rows and matching right rows
|right|	All right rows and matching left rows
|outer|	All keys from both datasets

**Important:** merge() can produce more rows than expected if a key appears multiple times in both datasets. Check whether your join keys are unique when you expect one record per customer.


**Real-world use case:** Combining CRM customer data, order history, marketing campaigns, payment records, and insurance policy details.

### 8. Example 6 — Work with dates, calculate monthly sales, and export results
* Time-series analysis


**Problem statement:** A retail business has daily transactions. Convert the date column into a proper date format, extract the month, calculate monthly revenue, and save the result to a CSV file.

```python

import pandas as pd

# Step 1: Create transaction data
data = {
    "Order_Date": [
        "2026-01-05",
        "2026-01-15",
        "2026-02-03",
        "2026-02-20",
        "2026-03-10"
    ],
    "Product": [
        "Laptop", "Mouse", "Laptop", "Keyboard", "Mouse"
    ],
    "Sales": [60000, 1500, 65000, 3000, 2000]
}

df = pd.DataFrame(data)

# Step 2: Convert strings into datetime values
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Step 3: Extract month
df["Month"] = df["Order_Date"].dt.month

# Step 4: Extract month name
df["Month_Name"] = df["Order_Date"].dt.strftime("%B")

# Step 5: Calculate monthly revenue
monthly_sales = df.groupby("Month_Name")["Sales"].sum()

print(monthly_sales)

# Step 6: Calculate total revenue
print("Total revenue:", df["Sales"].sum())

# Step 7: Export the original dataset to CSV
df.to_csv("sales_data.csv", index=False)

# Step 8: Export monthly report
monthly_sales.to_csv("monthly_sales.csv")

# Step 9: Read the saved dataset back
loaded_df = pd.read_csv("sales_data.csv")
print(loaded_df.head())
```
