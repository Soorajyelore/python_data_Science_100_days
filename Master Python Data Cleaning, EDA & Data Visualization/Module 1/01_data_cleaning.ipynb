## Exercise 1: Handle missing values
### Problem statement:

#### You are working as a Data Scientist for an insurance company. You receive customer data containing missing age, salary and city values.

## Your task is to:
1. Identify missing values.
2. Calculate missing-value percentages.
3. Replace numerical missing values using median.
4. Replace categorical missing values using mode.

### Step 1: Create sample dataset

```python
import pandas as pd
import numpy as np

data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106],
    "Age": [25, 30, np.nan, 45, 50, np.nan],
    "Salary": [30000, 40000, 50000, np.nan, 70000, 80000],
    "City": ["Pune", "Mumbai", np.nan, "Nagpur", "Pune", np.nan]
}

df = pd.DataFrame(data)

print(df)
```

## Step 2: Identify missing values

```python
print(df.isnull().sum())
```

Explanation:
- isnull() checks whether values are missing.
- sum() counts missing values in each column.

## Step 3: Calculate missing percentage

```python
missing_percentage = (
    
    df.isnull().sum() / len(df)

) * 100

print(missing_percentage)
```

Why?

This tells us what percentage of each column contains missing values.

### Step 4: Fill numerical missing values

```python
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

print(df)
```

```python
df["Salary"] = df["Salary"].fillna(
    df["Salary"].median()
)

df
```

Why median?

Median is less sensitive to extreme values than the mean.

# Step 5: Fill categorical missing values

```python
df["City"] = df["City"].fillna(
    df["City"].mode()[0]
)

print(df)
```

Why mode?

Mode returns the most frequently occurring category.

Expected result: No missing values should remain in these three columns.

## Exercise 3: Handle outliers using IQR
### Problem statement:

An insurance company has collected customer salary information. Some records contain unusually high salaries.

You need to identify and handle outliers using the Interquartile Range method.

```python
import pandas as pd

data = {
    "Salary": [
        25000, 30000, 35000, 40000,
        45000, 50000, 55000, 60000,
        65000, 1000000
    ]
}

df = pd.DataFrame(data)

```

```python
# Step 1: Calculate Q1 and Q3
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)


```

```python
# Step 2: Calculate IQR
IQR = Q3 - Q1

```

```python
# Step 3: Calculate boundaries
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

```

```python
# Step 4: Identify outliers
outliers = df[
    (df["Salary"] < lower_limit) |
    (df["Salary"] > upper_limit)
]

print("Outliers:")
print(outliers)
```

```python

# Step 5: Cap outliers
df["Salary_Capped"] = df["Salary"].clip(
    lower=lower_limit,
    upper=upper_limit
)

print(df)
```

Step-by-step explanation:

- Q1: 25th percentile.
- Q3: 75th percentile.
- IQR: Difference between Q3 and Q1.
- Lower limit: Q1 − 1.5 × IQR.
- Upper limit: Q3 + 1.5 × IQR.
- Values outside these limits are flagged as potential outliers.


Important: An outlier is not automatically an error. Investigate the business context before removing or capping it.

## Exercise 5: Handle inconsistent categorical values
#### Problem statement:

A retail company has customer records with inconsistent city and gender values.

For example:
- Pune
- pune
- PUNE
- Mumbai
- mumbai


Your task is to standardize these values.

```python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106],
    "City": [
        "Pune", "pune", "PUNE",
        "Mumbai", "mumbai", "MUMBAI"
    ],
    "Gender": [
        "Male", "male", "M",
        "Female", "female", "F"
    ]
}

df = pd.DataFrame(data)

# Step 1: Remove extra spaces
df["City"] = df["City"].str.strip()
df["Gender"] = df["Gender"].str.strip()

# Step 2: Standardize city
df["City"] = df["City"].str.title()

# Step 3: Standardize gender
gender_mapping = {
    "Male": "Male",
    "male": "Male",
    "M": "Male",
    "Female": "Female",
    "female": "Female",
    "F": "Female"
}

df["Gender"] = df["Gender"].map(gender_mapping)

# Step 4: Display cleaned data
print(df)
```

What is happening?

1. str.strip() removes unwanted spaces.
2. str.title() converts the first character of each word to uppercase.
3. Dictionary mapping converts different representations into consistent categories.
4. map() applies the mapping to every value.

## Exercise 6: Handle invalid and negative values
## Problem statement:

A healthcare company receives patient records. Some records have invalid ages, negative blood pressure readings and impossible BMI values.

Your task is to identify invalid values and replace them with missing values.

```python
import pandas as pd
import numpy as np

data = {
    "Patient_ID": [1, 2, 3, 4, 5],
    "Age": [25, 30, -5, 150, 45],
    "BMI": [22, 28, -10, 250, 32],
    "BloodPressure": [120, 130, -20, 140, 110]
}

df = pd.DataFrame(data)

# Step 1: Identify invalid ages
df.loc[
    (df["Age"] < 0) | (df["Age"] > 120),
    "Age"
] = np.nan

# Step 2: Identify invalid BMI
df.loc[
    (df["BMI"] <= 0) | (df["BMI"] > 100),
    "BMI"
] = np.nan

# Step 3: Identify invalid blood pressure
df.loc[
    (df["BloodPressure"] <= 0) |
    (df["BloodPressure"] > 300),
    "BloodPressure"
] = np.nan

# Step 4: Check invalid values
print(df)

# Step 5: Fill missing values
df = df.fillna(df.median(numeric_only=True))

print(df)
```

Important: These are illustrative validation boundaries, not clinical diagnostic thresholds. Real healthcare projects must use medically approved validation rules.

### Exercise 7: Feature encoding and feature scaling
#### Problem statement:

You are preparing customer data for a machine learning model.

Your dataset contains:

- Age
- Salary
- Gender
- City

Your task is to encode categorical variables and scale numerical features.

```python
import pandas as pd

from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    OneHotEncoder,
    LabelEncoder
)

data = {
    "Age": [25, 30, 35, 40, 45],
    "Salary": [30000, 40000, 50000, 60000, 70000],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "City": ["Pune", "Mumbai", "Nagpur", "Pune", "Mumbai"]
}

df = pd.DataFrame(data)

# Step 1: One-hot encode City
df = pd.get_dummies(
    df,
    columns=["City"],
    dtype=int
)

# Step 2: Label encode Gender
le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])

# Step 3: Standard Scaling
scaler = StandardScaler()

df[["Age", "Salary"]] = scaler.fit_transform(
    df[["Age", "Salary"]]
)

print(df)
```

Difference between scaling methods:

|Method	|Working	|Use case
:-------|-----------|-------------------
StandardScaler|	Mean = 0, standard deviation = 1	|Logistic regression, SVM
MinMaxScaler|	Usually scales between 0 and 1	|Neural networks, distance-based models
RobustScaler|	Uses median and IQR	|Data containing outliers

For real ML projects, fit the encoder and scaler on training data only, then transform test data. This prevents data leakage. Also, prefer OneHotEncoder for nominal categories such as gender and city rather than treating them as ordered labels.
