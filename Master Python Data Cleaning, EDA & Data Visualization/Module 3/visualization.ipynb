# Exercise 15: Histogram – Understand data distribution
- Problem statement:

An insurance company wants to understand the distribution of customer ages.

Create a histogram to identify:
- Most common age group.
- Distribution of customer ages.
- Whether the data is skewed.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Age": [
        22, 25, 28, 30, 32, 35,
        38, 40, 42, 45, 48, 50,
        55, 60, 65, 70, 75
    ]
}

df = pd.DataFrame(data)

# Step 1: Create histogram
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=8,
    kde=True
)

# Step 2: Add title and labels
plt.title("Customer Age Distribution")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")

# Step 3: Display chart
plt.show()
```

Step-by-step explanation:

1. plt.figure() defines the chart size.
2. sns.histplot() displays the frequency distribution.
3. bins=8 divides the age range into eight intervals.
4. kde=True adds a smoothed density curve.
5. plt.show() displays the visualization.

Interview question: What is the difference between a histogram and a bar chart?

|Histogram	|Bar chart
:------------|----------------
Shows numerical distribution|	Compares categories
Bars usually touch|	Bars usually have gaps
Example: Age distribution|	Example: Sales by city

### Exercise 16: Bar chart – Sales by city

- Problem statement:

A retail company wants to compare total sales across different cities.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "City": ["Pune", "Mumbai", "Nagpur", "Delhi", "Bangalore"],
    "Sales": [50000, 80000, 30000, 90000, 70000]
}

df = pd.DataFrame(data)

# Step 1: Create figure
plt.figure(figsize=(9, 5))

# Step 2: Create bar chart
sns.barplot(
    data=df,
    x="City",
    y="Sales"
)

# Step 3: Add chart details
plt.title("City-Wise Sales Performance")
plt.xlabel("City")
plt.ylabel("Total Sales")

# Step 4: Display chart
plt.show()
```

What can we understand?

- Which city has the highest sales?
- Which city has the lowest sales?
- Which market needs more attention?

For this dataset, Delhi has the highest sales and Nagpur has the lowest.

## Exercise 17: Box plot – Detect outliers visually
- Problem statement:

An insurance company wants to identify unusual salary observations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Salary": [
        20000, 25000, 30000, 35000,
        40000, 45000, 50000, 55000,
        60000, 900000
    ]
}

df = pd.DataFrame(data)

# Step 1: Create box plot
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Salary"
)

# Step 2: Add title
plt.title("Salary Outlier Detection")

# Step 3: Display
plt.show()
```

# How to interpret a box plot:

- Middle line: Median.
- Lower edge: Q1.
- Upper edge: Q3.
- Box: Interquartile range.
- Whiskers: Typical range according to the box-plot rule.
- Separate points: Potential outliers.

## Exercise 18: Scatter plot – Identify relationships

Problem statement:

A company wants to understand whether advertising expenditure affects sales revenue.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Advertising": [1000, 2000, 3000, 4000, 5000, 6000],
    "Sales": [10000, 15000, 18000, 25000, 30000, 35000]
}

df = pd.DataFrame(data)

# Step 1: Create scatter plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Advertising",
    y="Sales",
    s=100
)

# Step 2: Add regression line
sns.regplot(
    data=df,
    x="Advertising",
    y="Sales",
    scatter=False
)

# Step 3: Add labels
plt.title("Advertising vs Sales")
plt.xlabel("Advertising Expenditure")
plt.ylabel("Sales Revenue")

plt.show()
```

Explanation:

- X-axis represents advertising expenditure.
- Y-axis represents sales revenue.
- Each point represents one observation.
- Regression line shows the general linear trend.

If the points generally move upward, it suggests a positive association.

### Exercise 19: Correlation heatmap
- Problem statement:

A healthcare company wants to understand the relationships among Age, BMI, Glucose, Insulin and Blood Pressure.

Create a correlation heatmap.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Age": [25, 30, 35, 40, 45, 50],
    "BMI": [22, 24, 26, 28, 30, 32],
    "Glucose": [90, 100, 110, 120, 140, 160],
    "Insulin": [80, 90, 100, 110, 130, 150],
    "BloodPressure": [70, 75, 80, 85, 90, 95]
}

df = pd.DataFrame(data)

# Step 1: Calculate correlation
corr = df.corr()

# Step 2: Create heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

# Step 3: Add title
plt.title("Healthcare Feature Correlation")

plt.show()
```

## Understanding the heatmap:
|Correlation	|Interpretation
:---------------|------------------------
+1.0|	Perfect positive linear correlation
+0.7|	Strong positive correlation
0.0	|   No linear correlation
-0.7|	Strong negative correlation
-1.0|	Perfect negative linear correlation

annot=True displays correlation values inside the cells.

cmap="coolwarm" defines the color palette.

fmt=".2f" displays two decimal places.

### Exercise 20: Build a complete EDA dashboard
##### Problem statement:

You are working as a Data Scientist for an e-commerce company.

Management wants one dashboard containing:
1. Total revenue.
2. Average order value.
3. Total orders.
4. Revenue by city.
5. Revenue by product.
6. Monthly revenue.
7. Order value distribution.


This is a mini end-to-end EDA project.

Step 1: Create sample dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Order_ID": range(101, 111),

    "Order_Date": [
        "2026-01-05", "2026-01-10",
        "2026-01-15", "2026-02-01",
        "2026-02-10", "2026-02-15",
        "2026-03-01", "2026-03-05",
        "2026-03-10", "2026-03-15"
    ],

    "City": [
        "Pune", "Mumbai", "Delhi",
        "Pune", "Mumbai", "Delhi",
        "Pune", "Mumbai", "Delhi", "Pune"
    ],

    "Product": [
        "Laptop", "Mobile", "Tablet",
        "Laptop", "Mobile", "Tablet",
        "Laptop", "Mobile", "Tablet", "Laptop"
    ],

    "Quantity": [1, 2, 1, 1, 3, 2, 2, 1, 2, 1],

    "Sales": [
        50000, 30000, 15000,
        55000, 45000, 30000,
        60000, 25000, 35000, 52000
    ]
}

df = pd.DataFrame(data)
```

Step 2: Perform data cleaning

```python

# Check missing values
print(df.isnull().sum())

# Remove duplicates
df = df.drop_duplicates()

# Convert date
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"]
)

# Check data types
print(df.dtypes)

# Check dataset
print(df.head())
```

Step 3: Calculate business KPIs

```python
total_revenue = df["Sales"].sum()

average_order_value = df["Sales"].mean()

total_orders = df["Order_ID"].nunique()

total_quantity = df["Quantity"].sum()

print("Total Revenue:", total_revenue)

print("Average Order Value:", average_order_value)

print("Total Orders:", total_orders)

print("Total Quantity:", total_quantity)
```
