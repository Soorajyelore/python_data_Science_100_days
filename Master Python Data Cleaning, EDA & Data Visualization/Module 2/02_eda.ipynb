### Exercise 8: Understand dataset structure
- Problem statement:

You receive a customer dataset and need to understand:
- Number of rows and columns.
- Column names.
- Data types.
- Missing values.
- Statistical summary.

```python
import pandas as pd

df = pd.read_csv("customer_data.csv")

# Step 1: First five records
print(df.head())

# Step 2: Last five records
print(df.tail())

# Step 3: Dataset shape
print(df.shape)

# Step 4: Column names
print(df.columns)

# Step 5: Dataset information
print(df.info())

# Step 6: Statistical summary
print(df.describe())

# Step 7: Categorical summary
print(df.describe(include="object"))

# Step 8: Missing values
print(df.isnull().sum())
```

Explanation:
- head() shows the first records.
- tail() shows the last records.
- shape gives rows and columns.
- info() displays data types and non-null counts.
- describe() calculates statistical summaries.
- describe(include="object") summarizes categorical columns.

### Exercise 9: Univariate analysis
###### Problem statement:

An insurance company wants to understand the distribution of customer ages.

You need to calculate:
- Mean
- Median
- Mode
- Minimum
- Maximum
- Standard deviation
- Frequency distribution

```python
import pandas as pd

data = {
    "Age": [22, 25, 30, 35, 25, 40, 45, 50, 30, 25]
}

df = pd.DataFrame(data)

# Step 1: Mean
print("Mean:", df["Age"].mean())

# Step 2: Median
print("Median:", df["Age"].median())

# Step 3: Mode
print("Mode:", df["Age"].mode().tolist())

# Step 4: Minimum and maximum
print("Minimum:", df["Age"].min())
print("Maximum:", df["Age"].max())

# Step 5: Standard deviation
print("Standard Deviation:", df["Age"].std())

# Step 6: Frequency distribution
print(df["Age"].value_counts().sort_index())
```

Why is univariate analysis important?

It helps us understand one variable at a time and identify skewness, unusual values and common observations.

### Exercise 10: Bivariate analysis


Problem statement:

A retail company wants to understand whether customer age and salary are related.

```python
import pandas as pd

data = {
    "Age": [22, 25, 30, 35, 40, 45],
    "Salary": [20000, 25000, 35000, 40000, 50000, 60000]
}

df = pd.DataFrame(data)

# Step 1: Calculate correlation
correlation = df["Age"].corr(df["Salary"])

print("Correlation:", correlation)

# Step 2: Calculate covariance
covariance = df["Age"].cov(df["Salary"])

print("Covariance:", covariance)

# Step 3: Group salary by age
print(
    df.groupby("Age")["Salary"].mean()
)
```

Interpretation:
- Positive correlation: Both variables tend to increase together.
- Negative correlation: One tends to increase as the other decreases.
- Near-zero correlation: Little or no linear relationship.


Correlation does not prove causation.

### Exercise 11: GroupBy and aggregation

Problem statement:

A retail company wants to calculate total sales, average sales and number of transactions for each city.

```python
import pandas as pd

data = {
    "City": [
        "Pune", "Mumbai", "Nagpur",
        "Pune", "Mumbai", "Nagpur"
    ],
    "Sales": [10000, 20000, 15000, 12000, 25000, 18000],
    "Quantity": [2, 3, 1, 4, 5, 2]
}

df = pd.DataFrame(data)

# Step 1: Total sales by city
total_sales = df.groupby("City")["Sales"].sum()

print(total_sales)

# Step 2: Average sales
average_sales = df.groupby("City")["Sales"].mean()

print(average_sales)

# Step 3: Number of transactions
transactions = df.groupby("City")["Sales"].count()

print(transactions)

# Step 4: Multiple aggregations
result = df.groupby("City").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Transactions=("Sales", "count")
)

print(result)
```

Business use case: Identify which city contributes the highest revenue and where sales performance needs improvement.

## Exercise 12: Correlation analysis

Problem statement:

You are analyzing healthcare data and want to identify which features have a strong relationship with diabetes.

```python
import pandas as pd

data = {
    "Age": [25, 30, 35, 40, 45, 50],
    "Glucose": [90, 100, 110, 130, 150, 170],
    "BMI": [22, 24, 26, 28, 32, 35],
    "Diabetes": [0, 0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)

# Step 1: Calculate correlation matrix
correlation_matrix = df.corr()

print(correlation_matrix)

# Step 2: Correlation with target
target_correlation = correlation_matrix["Diabetes"]

print(target_correlation.sort_values(ascending=False))

# Step 3: Strongest absolute correlations
print(
    target_correlation.drop("Diabetes")
    .abs()
    .sort_values(ascending=False)
)
```

Why is this useful?

It helps us understand which numerical features are associated with the target variable.

For real diabetes modeling, use a sufficiently large, representative dataset and do not interpret correlation as a causal or clinical relationship.

### Exercise 13: Customer segmentation using RFM analysis


Problem statement:

You work for a retail company. Management wants to segment customers based on purchasing behavior.

RFM means:
- Recency: How recently did the customer purchase?
- Frequency: How many purchases did the customer make?
- Monetary: How much money did the customer spend?

```python
import pandas as pd

data = {
    "Customer_ID": [101, 101, 102, 102, 103, 103, 104],
    "Invoice_Date": [
        "2026-01-01", "2026-01-10",
        "2026-01-05", "2026-02-01",
        "2026-01-15", "2026-02-10",
        "2026-01-20"
    ],
    "Invoice_ID": [
        "I1", "I2", "I3", "I4",
        "I5", "I6", "I7"
    ],
    "Amount": [1000, 1500, 2000, 2500, 500, 700, 3000]
}

df = pd.DataFrame(data)

# Step 1: Convert invoice date
df["Invoice_Date"] = pd.to_datetime(df["Invoice_Date"])

# Step 2: Set analysis date
analysis_date = pd.Timestamp("2026-03-01")

# Step 3: Calculate RFM
rfm = df.groupby("Customer_ID").agg(
    Recency=("Invoice_Date",
             lambda x: (analysis_date - x.max()).days),

    Frequency=("Invoice_ID", "nunique"),

    Monetary=("Amount", "sum")
)

print(rfm)

# Step 4: Create RFM scores
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Step 5: Calculate overall score
rfm["RFM_Score"] = (
    rfm["R_Score"] +
    rfm["F_Score"] +
    rfm["M_Score"]
)

print(rfm)
```

Explanation:

- groupby() groups records by customer.
- max() finds the latest transaction date.
- nunique() counts unique invoices.
- sum() calculates total spending.
- qcut() creates quantile-based customer scores.

This small dataset is for practicing the method; real segmentation needs enough customers and appropriate treatment of ties.

## Exercise 14: Time-series EDA

Problem statement:

An e-commerce company wants to understand monthly revenue trends.

```python
import pandas as pd

data = {
    "Order_Date": [
        "2026-01-05", "2026-01-15",
        "2026-02-05", "2026-02-20",
        "2026-03-10", "2026-03-25"
    ],
    "Sales": [10000, 15000, 20000, 18000, 25000, 30000]
}

df = pd.DataFrame(data)

# Step 1: Convert date
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Step 2: Set date as index
df = df.set_index("Order_Date")

# Step 3: Resample monthly
monthly_sales = df["Sales"].resample("ME").sum()

print(monthly_sales)

# Step 4: Calculate monthly growth
growth = monthly_sales.pct_change() * 100

print(growth)

# Step 5: Calculate rolling average
rolling_average = monthly_sales.rolling(
    window=2
).mean()

print(rolling_average)
```

##### Why do we use these methods?
|Method	|Purpose
:-------|---------------------
to_datetime()|	Converts dates
set_index()|	Sets date as index
resample("ME")|	Aggregates at month-end
pct_change()|	Calculates percentage change
rolling()|	Calculates moving averages
