# 1. What is NumPy, and why do we use it?
- NumPy stands for Numerical Python.
- It provides a powerful data structure called an ndarray (N-dimensional array), which stores and processes numerical data efficiently.
- For example, suppose you have 1,000 employee salaries and want to increase every salary by 10%.
Without NumPy, you might use a loop:

```python
salaries = [20000, 30000, 40000]

updated_salaries = []

for salary in salaries:
    updated_salaries.append(salary * 1.10)

print(updated_salaries)
```

# With NumPy:

```python
import numpy as np

salaries = np.array([20000, 30000, 40000])

updated_salaries = salaries * 1.10

print(updated_salaries)
```

Why use NumPy?
- Fast mathematical calculations on large numerical arrays.
- Perform operations on entire arrays without writing explicit Python loops.
- Calculate mean, median, standard deviation, minimum, and maximum.
- Work with matrices and multidimensional data.
- Prepare numerical features for machine learning algorithms.
- Generate random numbers and sample datasets for testing models.

# 4. Example 1 — Create NumPy arrays
- **Problem statement:** Create arrays containing employee ages, monthly salaries, and a 3 × 3 matrix. Identify the dimensions and shapes.

```python
import numpy as np

# 1. One-dimensional array
ages = np.array([22, 25, 30, 35, 40])

# 2. Two-dimensional array
salaries = np.array([
    [20000, 25000],
    [30000, 35000],
    [40000, 45000]
])

# 3. Create arrays using NumPy functions
zeros = np.zeros(5)
ones = np.ones(4)
numbers = np.arange(1, 11, 2)
values = np.linspace(0, 10, 5)

print("Ages:", ages)
print("Salaries:\n", salaries)
print("Zeros:", zeros)
print("Ones:", ones)
print("Numbers:", numbers)
print("Values:", values)

print("Age dimensions:", ages.ndim)
print("Salary dimensions:", salaries.ndim)
print("Salary shape:", salaries.shape)
print("Salary size:", salaries.size)
```

**Step-by-step explanation**
1. np.array() converts a Python list into a NumPy array.
2. np.zeros(5) creates five zeros; np.ones(4) creates four ones.
3. np.arange(1, 11, 2) generates numbers starting at 1, increasing by 2, and stopping before 11.
4. np.linspace(0, 10, 5) generates five equally spaced values from 0 to 10.
5. .ndim returns the number of dimensions.
6. .shape returns the length of each dimension; (3, 2) means 3 rows and 2 columns.
7. .size returns the total number of elements.


Why is this useful? You need to understand array dimensions and shapes before using NumPy with pandas datasets and machine learning models.

#### 5. Example 2 — Indexing, slicing, and filtering
- Problem statement: Given monthly sales figures, find the first month's sales, extract the first three months, and identify months with sales greater than ₹30,000.

```python
import numpy as np

sales = np.array([15000, 22000, 28000, 35000, 42000, 50000])

# Access individual elements
print("First month:", sales[0])
print("Third month:", sales[2])
print("Last month:", sales[-1])

# Extract a portion of the array
print("First three months:", sales[:3])
print("Months 3 to 5:", sales[2:5])

# Filter values
high_sales = sales[sales > 30000]

print("Sales above 30000:", high_sales)
print("Number of high-sales months:", np.sum(sales > 30000))
```

**What is happening?**
- sales[0] accesses the first element because Python indexing starts at zero.
- sales[-1] accesses the last element.
- sales[:3] selects positions 0, 1, and 2.
- sales[2:5] selects positions 2, 3, and 4; the ending index is excluded.
- sales > 30000 creates a Boolean array.
- sales[sales > 30000] uses that Boolean array to select matching values.
- np.sum(sales > 30000) counts the True values because NumPy treats True as 1 and False as 0 in this calculation.

Why is this useful? Filtering is fundamental to finding high-value transactions, identifying outliers, and selecting records that meet business conditions.

## 6. Example 3 — Mathematical operations and broadcasting
- Problem statement: A shop has product prices and wants to calculate discounted prices, add tax, and calculate total revenue.

```python
import numpy as np

prices = np.array([100, 200, 300, 400, 500])
quantities = np.array([2, 3, 1, 4, 2])

# Apply a 10% discount
discounted_prices = prices * 0.90

# Add 18% tax to discounted prices
final_prices = discounted_prices * 1.18

# Calculate revenue for each product
revenue = final_prices * quantities

print("Original prices:", prices)
print("Discounted prices:", discounted_prices)
print("Final prices with tax:", final_prices)
print("Revenue per product:", revenue)
print("Total revenue:", np.sum(revenue))
```

Step-by-step explanation
1. prices * 0.90 applies a 10% discount to every price.
2. discounted_prices * 1.18 applies 18% tax to every discounted price.
3. final_prices * quantities calculates each product's revenue.
4. np.sum(revenue) adds the revenue values to produce total revenue.


* The key concept is vectorization: NumPy performs operations on entire arrays without requiring an explicit Python loop.


* This example also demonstrates broadcasting. The scalar 0.90 is applied to every element in the prices array.


Other common operations include:

```python
a = np.array([10, 20, 30])

print(a + 5)    # [15 25 35]
print(a - 5)    # [ 5 15 25]
print(a * 2)    # [20 40 60]
print(a / 10)   # [1. 2. 3.]
print(a ** 2)   # [100 400 900]
```

Why is this useful? It is useful for financial calculations, feature transformations, percentage changes, and calculations across large datasets.

# 7. Example 4 — Statistical functions in NumPy
Problem statement: An HR department wants to analyze employee salaries. Calculate the average salary, median salary, minimum, maximum, standard deviation, and salary range.

```python
import numpy as np

salaries = np.array([
    25000, 30000, 35000, 40000, 45000,
    50000, 55000, 60000
])

print("Mean salary:", np.mean(salaries))
print("Median salary:", np.median(salaries))
print("Minimum salary:", np.min(salaries))
print("Maximum salary:", np.max(salaries))
print("Standard deviation:", np.std(salaries))
print("Salary range:", np.ptp(salaries))
print("Total salary:", np.sum(salaries))
```

# Understanding each function

|Function	|Meaning	|Why use it? |
:-----------|-----------|------------|
|np.mean()	| Arithmetic average	|Find average salary or marks
|np.median()|	Middle value after sorting	|Understand the center when outliers exist
|np.min()	|Smallest value	|Find the lowest salary
|np.max()	|Largest value	|Find the highest salary
|np.std()	|Standard deviation	|Measure how spread out values are
|np.ptp()	|Maximum minus minimum	|Calculate the range
|np.sum()	|Sum of values	|Calculate totals

By default, np.std() calculates the population standard deviation (ddof=0). 

For a sample standard deviation, use np.std(salaries, ddof=1).

Why is this useful? Statistical functions are essential for exploratory data analysis (EDA), reporting, detecting unusual values, and understanding distributions before machine learning.

# 8. Example 5 — Reshaping, sorting, and matrix operations
- **Problem statement:** A teacher has 12 marks and wants to organize them into 3 students with 4 subject marks each. Then calculate each student's total and average marks.

```python
import numpy as np

marks = np.array([
    80, 75, 90, 85,
    60, 70, 65, 75,
    95, 90, 85, 100
])

# Reshape into 3 rows and 4 columns
marks_matrix = marks.reshape(3, 4)

print("Marks matrix:\n", marks_matrix)

# Calculate totals and averages for each student
totals = np.sum(marks_matrix, axis=1)
averages = np.mean(marks_matrix, axis=1)

print("Total marks:", totals)
print("Average marks:", averages)

# Calculate average marks for each subject
subject_averages = np.mean(marks_matrix, axis=0)

print("Subject averages:", subject_averages)

# Sort all original marks
print("Sorted marks:", np.sort(marks))
```

# Important concepts
- reshape(3, 4) organizes 12 elements into 3 rows and 4 columns without changing the data.
- axis=1 calculates across columns within each row. Here, it calculates each student's total or average.
- axis=0 calculates down the rows for each column. Here, it calculates each subject's average.
- np.sort() sorts values in ascending order.

##### For a 2D array
### axis=0

* Move down the rows to calculate a result for each column.

* axis=1

### Move across the columns to calculate a result for each row.

# Bonus: Matrix multiplication
# Matrix operations are important in linear algebra and machine learning.

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("Element-wise multiplication:\n", A * B)
print("Matrix multiplication:\n", A @ B)
```

* A * B multiplies matching elements, whereas A @ B performs matrix multiplication using row-by-column calculations.
* Why is this useful? Reshaping is used to prepare data for algorithms, and matrix multiplication is a foundational operation in regression, neural networks, and other machine learning models.

# 9. Example 6 — Random numbers and generating a dataset
* Problem statement: Generate marks for 10 students, calculate their average, and identify students who scored at least 75.

```python
import numpy as np

# Make results reproducible
np.random.seed(42)

# Generate 10 random marks from 0 to 100
marks = np.random.randint(0, 101, size=10)

print("Student marks:", marks)

# Statistical calculations
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))

# Find students meeting the target
passed_target = marks >= 75

print("Boolean result:", passed_target)
print("Marks >= 75:", marks[passed_target])
print("Number scoring >= 75:", np.sum(passed_target))
```

**Explanation**
1. np.random.seed(42) makes the random sequence reproducible in the same NumPy random-number setup.
2. np.random.randint(0, 101, size=10) generates 10 integers from 0 through 100, inclusive.
3. np.mean(), np.max(), and np.min() summarize the generated marks.
4. marks >= 75 creates a Boolean mask.
5. marks[passed_target] selects only marks meeting the target.

**Why is this useful? Random data is useful for practice datasets, simulations, testing code, and generating sample inputs for machine learning exercises. For newer code, NumPy also offers np.random.default_rng().**

# Problem 1 — Sales data analysis
Problem statement: A store records sales for seven days. Calculate total sales, average sales, the highest-sales day, and all days with sales greater than ₹5,000.

```python

import numpy as np

sales = np.array([3000, 4500, 6000, 2500, 7000, 8000, 4000])

# Calculate summary statistics
total_sales = np.sum(sales)
average_sales = np.mean(sales)
highest_sales = np.max(sales)
lowest_sales = np.min(sales)

# Find the index of the highest sales
highest_day_index = np.argmax(sales)

# Find days with sales above 5000
high_sales_days = np.where(sales > 5000)[0]

print("Total sales:", total_sales)
print("Average sales:", average_sales)
print("Highest sales:", highest_sales)
print("Lowest sales:", lowest_sales)
print("Highest sales day (day number):", highest_day_index + 1)
print("Days with sales above 5000:", high_sales_days + 1)
```

### How the solution works
- np.sum(sales) adds all seven values.
- np.argmax(sales) returns the index of the highest value.
- Adding 1 converts the zero-based index into a day number from 1 to 7.
- np.where(sales > 5000)[0] returns the indices where the condition is true.


Expected key results: total sales = ₹35,000, average sales = ₹5,000, highest sales = ₹8,000 on day 6.

## Problem 2 — Student performance analysis

Problem statement: Three students have marks in Mathematics, Science, and English. Calculate each student's total and average, find the highest-performing student, and calculate the average for each subject.

```python

import numpy as np

# Rows represent students; columns represent subjects
marks = np.array([
    [85, 90, 80],
    [70, 75, 80],
    [95, 92, 98]
])

student_totals = np.sum(marks, axis=1)
student_averages = np.mean(marks, axis=1)
subject_averages = np.mean(marks, axis=0)

best_student_index = np.argmax(student_averages)

print("Student totals:", student_totals)
print("Student averages:", student_averages)
print("Subject averages:", subject_averages)
print("Best-performing student:", best_student_index + 1)
```

* Key lesson: axis=1 calculates each student's result, while axis=0 calculates each subject's result.


# Problem 3 — Employee salary and bonus analysis
* Problem statement: An HR department stores employee salaries in a NumPy array. Give employees earning below ₹30,000 a 15% raise and employees  * * * earning ₹30,000 or more a 10% raise. Calculate the updated payroll.

```python

import numpy as np

salaries = np.array([22000, 28000, 35000, 45000, 25000])

# Create a copy so original salaries remain unchanged
updated_salaries = salaries.astype(float).copy()

# Apply raises using Boolean indexing
low_salary = salaries < 30000
high_salary = salaries >= 30000

updated_salaries[low_salary] *= 1.15
updated_salaries[high_salary] *= 1.10

print("Original salaries:", salaries)
print("Updated salaries:", updated_salaries)
print("Original payroll:", np.sum(salaries))
print("Updated payroll:", np.sum(updated_salaries))
print("Additional payroll cost:",
      np.sum(updated_salaries) - np.sum(salaries))
```
